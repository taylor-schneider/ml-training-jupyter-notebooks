#### Hypothesis Testing

As the name suggests, hypothesis testing is a statistical process by which a hypothesis is defined and then accepter or rejected based on the data.

Over the years many different fields have adapted the process to their respective disipline. Doctors have conducted trials to determine whether a drug had a particular effect, Marketers have studdied whether advertisements affected purchases within key demographics, Political Poll Conductors have performed studdies to understand the views of the voter population, the list goes on.

Regardless of the field of study, the hypothesis tests all revolve around probability distributions (and the concept of likelihood which we will cover later). Depending on how the hypothesis is phrased (i.e. how we draw a conclusion) and what data is available (i.e. how big is the sample, how much do we know about ground truth) there are a number of different tests available. 

The most common tests one will encounter include $z$, $t$, $\chi^2$, F-ratio tests which we aim to cover below.

##### Z-Test & Z-Score

The purpose of a z-test is to make an inference about the relationship between two sets of data. Specifically, the tests allows us to compare the means of two sets which are assumed to be normally distributed.

**Note**: It is common to refer to the sets as samples, or a sample and a population. This language is derived from the applied sciences that used the techniques for experimentation and traditional hypothesis evaluations. For example, biological testing or pharmesutical testing of whether a drug had a significant impact on a test subject during a clinical trial.

There are two variations for how this test is applied. 

1. One Sample - Does a sample mean differ from the theoretical distribution's mean
2. Two Sample - Does a sample mean differ from another sample mean

The calculations for the test relies on the Cumulative Density Functions (CDFs) of the two sets in question. The underlying mathematical nature of the CDFs allows us to frame comparisons in several ways
* in terms of direct equality (i.e. this mean is not equal to that one)
* in terms of general comparabilitiy (i.e. this mean is greater than or less than that one)

**Note**: This assumption of normality mentioned previously is a major requirement. It is an established mathematical principal that all normal distributions can be transformed (i.e. scaled, normalized, standardized) into a standard normal distribution. By stanrdizing the two samples we are effectively able to make "apples to apples" comparions and justify our inference.

Additionally, The logic for the comparison is based on the concept of likelihood. Generally speaking, likelihood is a relative measure for how common or uncommon we expect a paticular value is given that we know it's CDF. In the case of the z-test, we assess how likely the observed value from the first sample is, given that we assume the second sample's CDF is reposnsible for generating the obervation. If the observation is extremely unlikely, we infer that the CDFs of the two samples are materially different.

### Bayesian Statistics

#### Bayesian Inference

The Bayesian Inference framework is a class of Statistical Inference techniques and is so named due to it's fundamental reliance on the Bayesian Theorem.

As an inference framework, The Bayesian approach attemps make a statistical inference about an unobserved probability distribution. Another way of saying this would be that the approach attemps to draw a conclusion about an unobserved distribution. The inference, or conclusion is provided by a logical argument (often abstracted as a model or formula) which relies on some observation(s). And as the name suggests, this logical argument relies on Bayes Theorem.

##### Conventions For Inference

Many texts and disciplines have their own conventions and terminology associated with how the inference is defined and how it will be used. 

For example, in some cases, the conclusion being drawn, is framed as the acceptance or rejection of a hypothesis given some evidence. In other cases, the conclusion is framed as the estimation of a population distribution (or its defining parameter set) given a set of sample data. However this is an over simplification; the true conclusion would be whether or not a given estimate is accepted or rejected based on the given obeserved sample data. 

Additionally, we might go a step further to draw multiple conclusions and then choose the best one; in other words, select the estimate associated with the highes probability from the set of acceptable estimates. Later we will examine the methods used for this purpose.

We see that the comonality between all of these cases of inference are that we are trying to draw some conclusion (will we accept or refute) an objective claim about an unobserved probability distribution.

##### Basic Inference Process

In order to draw a conclusion (i.e. produce an inference) one typically performs the following activities:

* declares assumptions
* collects information or makes an observation (i.e. samples the population)
* analyzes the observations and information (sample set) to produce some summary or descriptive statistics
* defines a relationship between the sample distribution and population distribution (i.e. a model)
* estimates some quantity or characteristic of the population using the model
* articulates an argument for or against the acceptance of the estimate
* makes additional inferences based on the prior inference

##### Appication Of Bayes Theorem

With the bayesian framework the model is defined such that it relies on or utilizes the Bayes Formula which essentially acts as the model. 

Recall that the basic definition for Bayes Theorem is

$$ P(B|A) = \frac{P(B\cap A)}{P(A)} = \frac{P(A | B)P(B)}{P(A)} $$

**Note**: The fraction on the far right being expanded via a manipulation of the definition of conditional probability.

**Note**: A proof of the theorem can be found [here](https://byjus.com/maths/bayes-theorem/).

With this formula we are essentially saying that we will calculate the probability of some event $B$ occurring given, or based on the prior occurance of, some event $A$. This conditional probability is  denoted as $P(B|A)$.

**Note**: While there is a correlation assumed between $A$ and $B$ there is not a causal relationship assumed. That being said, the formula can model causal relationships if additional assumptions are made.

Where things get interesting is when we plug in the mathematical formulas for probability and perform algebraic manipulations to simplifiy and partially solve the equations.

##### Modeling Considerations & Causality

But what are these events, really? What are we actually modeling?

To be frank, the framework is quite flexible and the model could be used to describe a number of things. 

We could assess the probability you have a disease given a specific test restult for example. In this example, there is not necessarily a causal relationship implied between the two events being modeled.

We could also assess the probability that a sample set yields a particular point estimate given the population is assumed to have, or empirically dempostrates, a specific parameter or distribution. This example does imply a causal relationship as the sample statistics will converge towards the population statistics as more samples are selected.

Similarely we could asses the probability it will rain given that is has not rained in some time. Again, not necessarily a causal relationship, but also not clearly non-causal. The assessment of the causality is not directly addressed by Bayes Theorem and thus not the objective of the Bayesian Inference Framework.

> One reason why causality cannot be expressed by conditional probability (without further assumptions) is that you can turn $P(A|B)$ into $P(B|A)$ with Bayes theorem and causality does not go both ways (similar to the saying „correlation does not imply causation“).
>
> [stackexchange](https://stats.stackexchange.com/questions/454516/conditional-probability-and-causality)

It is also common to hear that the framework described as being able to "update the probability for a hypothesis as new information becomes available". In this example, we are assuming that $A$ and $B$ are effectively two samples of the same random process. They may or may not be independent, and the "new information" is that observed from the realization of $A$. In this situation, we "update" our prior probability by calculating the posterior. This "updated" probability thus provides a more informed prediction for the relization of $B$. Over time, with more trials/samples, the descrete statistics for the samples should converge with that of the population.

##### The Prior and Posterior
As this framework is a sequential process that inherently unfolds over time, there is a temporal aspect to the terminology that is used to describe its various elements. Specifically I am referring to the adjectives prior and posterior which are attached to terms in Bayes Theorem. The prior distribution, for example, refers to the probability distribution we attach to $A$. Then, after we perform the inference process, we end up with the probability distribution of (or at least related statistics for) $B$.

A point of contention, or at least scrutiny for Bayes Theorem has to do with the selection of a Prior. This is one of the assumptions made from the outset. There are several arguments commonly used to support the selection of a prior including:
* **Symmetry** - the structural mechanics of a system that logically results in equally probability being assigned to all possible outcomes
* **Axioms** - knowledge of the domain (i.e. valid possible values that the system can realize) which are known a priori due to the structural mechanics of the system can be used to define the prior
* **Empirical Evidence** - existing sample sets which are deemed valid may inform the selection of a prior
* **Subjectivity** - a choice based on luck or personal belief which results in a subjective and possibly irrational argument can be offered for the selection of a prior

Depending on how the prior is selected, statisticians and theorists make the case that the Bayesian Inference Framework and thus the inferences it generates are subjective. You will hear statements like "The Bayesian method allows one to attach subjective probabilities...". This is not always the case as we can model classical games of chance, which have an agreed upon and objective prior.

##### Conjugate and Non-Conjugate Priors

The term Conjugate prior was introduced by Howard Raiffa and Robert Schlaifer in *Applied Statistical Decision Theory* published in 1961. While the word has a specific meaning in this context, we can derive some intuition about the words and motivations for its selection by examining the use of the word in more abstract settings. For example in language we conjugate verbs to agree with the subject or tense. We can also conjugate, or take the conjugate of, algebraic expressions. In either case the term conjugate (verb or noun) alludes to some logical relationship between two things. These things are the exact same with the exception of a noted feature or characteristic. By changing that feature or characteristic we conjugate the term from one form to another. But these conjugated forms are still within the same "family" or logical set of forms.

With the Bayesian approach, we have the ability to specify prior and posterior in terms of conjugates.

* **Conjugate Priors** – When the prior and posterior distribution belong to the same family of distributions. For example the Gamma and Poisson distributions are both members of the Exponential family of distributions. Similarely we have Beta and Binomial being conjugates as well as different versions of the Normal distribution with respect to each other.

  The benefit of the cojugate relationship is that there is, generally speaking, a closed form algebraic way to transform the prior into the posterior. This provides a closed form analyitical solution to the inference problem.

* **Non-conjugate Priors** – When the prior and posterior distribution do not belong to the same family of distributions.

  In these cases it may not be possible to reach a closed form anlytical solution for the posterior. As such one would need to employ simulation methods to approximate the solution within a margin of error. This is a common issue found in optimization problems.

* **Un-informative prior** – When the prior is specifically selected so that it minimizes the amount of information that is captured and modeled by the prior. This is typically done to reduce bias by maximizing the influence of the information modled by the posterior.

##### Contingency Tables

In order to decide whether to accept or reject a Bayesian Inference, it is common to utilize a contingency table. This table organizes the set of mutually exclusive calculations that Beyes Theorem provides and allows us to quantify the probabilities associated with the inference being 

|       | B              | !B               | Total |
|:-----:|:--------------:|:----------------:|:-----:|
|  $A$  | $P(A \cap B)$  | $P(A \cap !B)$   |  $P(A)$ |
|  $!A$ | $P(!A \cap B)$ | $P(!A \cap !B)$  |  $P(!A)$ |
| Total | $P(B)$         | $P(!B)$          |  1    |

Now if we divide each cell with its coresponding total, via the definition of conditional probability, we arrive at the conditional probability.

$$ \frac{P(A \cap B)}{P(B)} = P(A|B) $$

$$ \frac{P(A \cap B)}{P(B)} = \frac{P(B \cap A)}{P(A)}  = P(B|A) $$

As such we can look at the probabilities of each set of outcomes and objectively accept or reject the coresponding inference. For example if $P(B|A) < P(!B|A)$ we would conclude that it's more probable that $B$ would not occur given $A$ as compared to $!B$.

A few simple examples of using this table can be found [here](https://www.wallstreetmojo.com/bayesian-inference/).

The posterior event $B$ is also treated as a random variable. So to is there a similarely a set of possible values for $b$ which can be realized for $B$.

**Note**: With hypothesis testing it is common to refer to the realization of the posterior as the observation. Thus the observational data is a realization of the random variable.

##### Connection With Likelihood

The Bayes Theorm has a strong connection with the concept of likelihood. However, the concept of likelihood was not developed until some ~300 years later in the 20th century. 

Looking at the basic definition for Bayes Theorem we have:

$$ P(B|A) = \frac{P(A | B)P(B)}{P(A)} $$

**Note**: The fraction on the far right being expanded via a manipulation of the definition of conditional probability.

Breaking this equation down into individual terms, we have the following:

* $ P(B|A)$ - This conditional probability is referred to as the **posterior**
* $ P(A | B)$ - This conditional probability is referred to as the **likelihood**
* $ P(B)$ - This marginal probability is sometimes referred to as the **evidence**
* $ P(A)$ - This marginal probability is referred to as the **prior**

We see that one of these terms is referred to as the likelihood. We will not get into the interpretation or symantics related to likelihood in this notabook.

## Degrees Of Probability

Bayes mentions this term in his posthumous publication. As it is repeated serveral times and given as part of his rationale and high level description of his theorem, I felt it impertative to understand the origin and meaning of the term.

I googled and found a helpful reference from a Swiss mathematician named Charles Benjamin, baron de Lubières. His publication was [translated](https://quod.lib.umich.edu/d/did/did2222.0000.983/--probability?rgn=main;view=fulltext;q1=Charles-Benjamin%2C+baron+de+Lubi%C3%A8res) from french to english by Daniel C. Weiner from Boston University. In 1765 Benjaman [published](https://artflsrv04.uchicago.edu/philologic4.7/encyclopedie0922/navigate/13/1556) *Probabilité*


In this paper he defines "degree of probability" as something similar to what we would call a "degree of certainty" today. He references "the Geometers" as conceiving that probability is a part while certainty is the whole. Through his diologue we see that degrees are defined in terms of odds. And that with the odds being 9:1, we have a degree of 9 for some particular event and a degree of one for the opposing event. Thus the two degrees of certainty are the odds associated with an event.

> Probability, every proposition considered in and of itself is true or false; but for us, it can be certain or uncertain; we can more or less perceive the relations which can exist between two ideas, or the agreement of one with the other, founded on certain conditions which link them, and which, when they are completely known to us, give us certainty of this truth, or of this proposition; but if we only understand a part of it, we then have only a simple probability , which seems that much more believable the greater the number of these conditions of which we are assured. It is these latter which form the degrees of probability, a correct opinion and an exact measurement of which would offer the pinnacle of wisdom and of prudence.

Geometers judged that their calculations could serve to evaluate these degrees of probability , at least to a certain point, and they had recourse to Logic, or the art of reasoning, to discover its principles and establish a theory for it. They regarded certainty as a whole and probabilities as parts of this whole. Consequently the correct degree of probability of a proposition was known to them exactly, when they could state and prove that this probability was worth a half, a quarter, or a third of this certainty. Often they were satisfied in supposing it; their calculations in and of themselves were no less correct; and these expressions, which at first can seem a bit bizarre, are no less meaningful. Examples taken from games, bets, or insurance will clarify them. Suppose someone comes and tells me that I have drawn the sum of ten thousand pounds in a lottery, I doubt the truth of this news. Someone who was there asks me what sum I would wager on the truth of that statement. I offer him half, which means that I only regard the probability of this news as a half-certainty; but if I had only offered a thousand pounds, it would have meant that I had nine times as much reason to believe in the truth of this news than not to believe in it. Or it would bring the probability to nine degrees, in the sense that certainty having ten, there would only lack one the addition of one more to give complete faith in the news.

He goes on to define some additional terms which I have seen referenced in a number of articles of the time:

> In ordinary usage, we call probable what has more than a half-certainty of credibility, or what surpasses this half considerably; and morally certain , what is virtually complete certainty. We speak here only of moral certainty, which coincides with mathematical certainty, although it is not susceptible to the same proofs. Moral evidence is thus, properly speaking, a probability so great that it is wise for a man to think and act, in cases where one has this certainty, as one should think and act, if one had a mathematical certainty of it.

## Interpretations of Probability

When we talk about the interpretaion of probability we are talking about an analysis of the notion of probability or the understanding of what probability measures. In other words we are asking the question "what does probability mean".

From the texts I have reviewed, there appears to be a concensus that the two main classifications for the interpretations of probability are either:

1. Objective - probability measures a feature of reality independend from knowledge or belief
1. Epistemological (subjective) - probability measures the degree of human knowledge or belief


### Objective Interpretations

An objective truth or observation is a set of information which exists independently of an individuals perception or beliefs. Objective arguments are seen as logical being based on independently verifiable facts and repeatable processes. An objective statistic is thus a mathematical quantity which was derived through objective facts and logical reasoning.

Objective inferences are those which are based on objective information. Additionally, the inferrential process is also conducted in a way that it is not influenced by any individual perceptions or beliefs.

Because the objective statistics and inferences are verifiable and irrifutable by third parties they are held in high regard as compared to the non-objective (i.e. subjective) alternatives.

A powerpoint on the history of objective Bayesian can be found [here](https://blogs.kent.ac.uk/jonw/files/2015/04/Berger2005.pdf).

### Epistemological or Subjective Interpretations

Subjective arguments are the opposite; they are based on feelings, opionions, or emotions and cannot be independently verified by repeatable processes.

There are two types of Subjectivism:

**Weak subjectivism**: In this interpretation, the term "subjective" is used in its weaker sense, meaning merely that probability encapsulates the rational beliefs of a subject. (Some people, such as myself, prefer to use the term "epistemic" for this concept, since it does not actually require subjectivity in the stronger sense.)


**Strong subjectivism**: In this interpretation, the term "subjective" is used in its stronger sense, meaning that weak subjectivism holds, and furthermore, the subject's belief lacks any outside "objective" justification (i.e., two or more different subjects could all hold different beliefs, and none would be considered more or less wrong than the others).

Bayesian Inference, for example, has both subjective and objective forms.

A great stackexchange article on the subject can be found [here](https://stats.stackexchange.com/questions/381825/objective-vs-subjective-bayesian-paradigms).

### (1838-1950) - Inverse Probability

Inverse Probability is a concept implied through Bayes' Theorem despite Bayes and Laplace not considering the topic. A helpful stackexchange article can be found [here](https://stats.stackexchange.com/questions/492966/what-exactly-does-the-term-inverse-probability-mean).

Feinberg (2006) notes that 

> the method of inverse probability ... flowed from (Baye's Theorem)...

And that until the middle of the twentieth century, the term referred inferrential methods which are now referred to as Bayesian.

Feinberg (2006) goes on to suggest that De Morgan coined the term in his writings and that he attributed its general form to Laplace's 1812 book despite Laplace never using the term in his 1774 paper or the 1812 book. He notes that De Morgan chose the term inverse

> because it involves inferring backwards from the data to the parameter 
or from effects to causes

Figuratively, the term inverse refers to the oposite of direction, position, order, or effect. Recall that the inverse of a function $y=f(x)$ is such that $x = f^{-1}(y)$. To invert a function, switch the positions of the variables and then rearrange the equation so that it returns values for the independent variables. Feinber's comments imply that De Morgan's understanding was that normal probabilities infer "forwards" from the parameter to the data; i.e. from cause to effect. 

Thinking this through if, a probability is an abstraction for a function; i.e. it is a function with certain characteristics. Given the conditional probability function $P(B|A)$, the inverse probability would be $P(A|B)$. 

Examining Bayes Theorem, we see that both of these terms are present. Thus Bayes Theorem allows for the calculation of inverse probabilities given regular probabilities..

But many argue that Bayes and Laplace were not aware of or intending for such concepts. Feinberg (2006) notes that this cause a debate between mathemeticians about the meaning of inverse probability and the meaning of probability in general.

> Mathematicians and philosophers continued to debate the meaning of probability
throughout the remainder of the nineteenth century. For example, Boole, Venn, Chrystal, and others argued that the inverse method was incompatible with objective probabilities, and they also critiqued the inverse method because of its use of indifference
prior distributions.

At first, the concept of inverse probability does not seem that controversial. Then thinking about classical problems like the rolling of dice, the calculation of probabilities of rolling a six given we just rolled a four, and then reversing the calculation seem straigt forward. 

However, it does start to become somewhat suspect when discussing various applied sciences. For example, when estimating a population parameter, the probability of an event is assumed conditional; the probability of an event is assumed to be conditional given some population distribution. But what would the inverse be? The probability of a distribution given an event? Basically, yes. But as we will discuss, this starts to "change the meaning" of probabilitiy. As such, people like Fisher argue that we should use the term liklihood instead to denote the inverse probabilities.

> Fisher’s work had a profound influence on two other young statisticians working in
Pearson’s laboratory at University College London: Jerzy Neyman, a Polish statistician
whose early work focused on experiments and sample surveys, and Egon Pearson, Karl
Pearson’s son. They found Fisher’s ideas on significance tests lacking in mathematical
detail and, together, they set out to extend and “complete” what he had done. In the
process, they developed the methods of hypothesis testing and confidence intervals that
were to revolutionize both the theory and the application of statistics. Although Fisher
often disagreed with them caustically and vigorously both orally and in print (e.g.,
see Fienberg and Tanur (61)), some amalgam of their approaches—referred to later as
“frequentist” methods—soon supplanted inverse probability. The mathematical foundation for these methods emerged from Russia and the axiomatic work of Kolmogorov
(93).
