# Transformations

In the context of vector transformations, a transformation is another word for a function which takes a vector as an input and returns a vector as an output, or the resultant output of such a vector function.

We can infer that the choice in terminology is related to a visualization of how in input vector is "transformed" or "mapped" to the output function.

## Linear Transformations

Linear transformations are a special type of transformation which satisfy two criteria. In lamens terms, these are:

1. The origin is not transformed; i.e. it does not move
2. All lines remain lines; i.e. vertical, horizontal, and diagonal lines within the input space remain lines and do not curve in the output space

Written mathematically, we would say that a tranformation $T$ is linear iff. the following equations hold for all vectors $v$ and $w$ and for all scalars $c$:

$$ T(v + w) = T(v) + T(w) $$

$$ T(cv) = cT(v) $$

## Projections

A projection is a linear transformation.

I believe the term projection was adopted through various applications of linear transformations. The general definition of a projection is an estimate based on a study or the presentation of an image on a surface. In the case of vectors, if we think about the resultant transformation as us presenting the vector on a surface; the surface being defined by the transformation.
