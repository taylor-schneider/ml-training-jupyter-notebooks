# Definition

The function is usually given in the following form:

$$ p_i = \frac{e^{z_i}}{\sum_j^M{e^{z_i}}} $$

We see that it is an exponential function that is then normalized by the summation or integral of the terms over the interval.

# Etymology

A helpful stackexchange can be found [here](https://math.stackexchange.com/questions/1888141/why-is-the-softmax-function-called-that-way).

Many infer that the softmax function was named in direct contract to the normal max function (i.e. hard max). The hardmax function will only inform which element is the maximum while the softmax will inform the relative rank of each element in the original set.

An alternate inference is that the function is differentiable or soft at all points in the function. This is analogus to the softplu function being a smooth, continuous, and differentiable sigmoidal function which can approximate the ReLU rectifier which is none of those things.

# Connection With Logistic Function

There are many articles claiming that the logistic function is a multidimensional generalization of the logistic function. I believe this is overly broad and misleading. For more information on the logistic function, see the [sigmoid funstions notebook](Sigmoid%20Functions.ipynb).

This claim is strictly true in only a very specifc circomstance, if $z=[0,x]$ as noted [here](https://deepai.org/machine-learning-glossary-and-terms/softmax-layer).

$$ \frac{e^{z_i}}{\sum_j^M{e^{z_i}}} = \frac{e^{z_x}}{e^{z_0} + e^{z_x}} = \frac{e^x}{e^0 + e^x} = \frac{e^x}{1 + e^x}$$

The similartiy is that the softmax function allows for classification into more than two mutually exclusive classes. This can then be thought of as a generalization of the logistic function which allows classification into two mutually exclusive classes. 

But, the two functions have different mathematical forms and different properties. They were developed during different time periods for different purposes. Additionally, the softmax function is not sigmoidal which is one of the main ientifying factors of the logistic regression.

One of the most notable differences is that the softmax function is normalized and thus all the values of the function sum to one.

The softmax function operates on a vector while the logistic function operates on a scalar.

### Boltzmann (1868) - Origin Of Distribution

Over several years, physicists were trying make calculations about the energy and states of atomic particles.

In 1868, Boltzman [published](https://gilles.montambaux.com/files/histoire-physique/Boltzmann-1868.pdf) *Studien über das Gleichgewicht der lebendigen Kraft zwischen bewegten materiellen Punkten* which roughly translates to “On the Relationship between the Second Fundamental Theorem of the Mechanical Theory of Heat and Probability Calculations Regarding the Conditions for Thermal Equilibrium".

Boltzmann put forward a law for the probability of finding an energy $H$ in some molecule being proportional to

$$e^{-\frac{H}{kT}}$$

The proportionaliy is dictated by a constant which givides the energy by total energy in a system.

### Bridle (1989) - Intorduction of term Softmax

I have seen many articles ([example](https://ai.stackexchange.com/questions/22426/which-paper-introduced-the-term-softmax)) citing Bridle as being responsible for intoducing the term softmax in his 1989 [publication](https://proceedings.neurips.cc/paper/1989/hash/0336dcbab05b9d5ad24f4333c7658a0e-Abstract.html) of *Training Stochastic Model Recognition Algorithms as Networks can Lead to Maximum Mutual Information Estimation of Parameters*.

$$ Q_j(x) = \frac{\exp(V_j(x))}{\sum_k{\exp(V_j(x))}} $$

Here, he refers to the softmax as being a normalized exponential. This is because the formula of the function exponentiates each element (i.e. raises $e$ to the power of that element) and then normalizes the result by dividing it by the sum of all the exponentiated terms.
