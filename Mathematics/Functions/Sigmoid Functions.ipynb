# Etymology

Sigmoid, or sigmoidal functions, derive their name from the greek letter Sigma. The letter takes on a third form (as opposed to the normal two cases: upper and lowercase) when occurring at the end of a word. The form is written $\varsigma$ and over time has been written as either a C-like shape or an S-like shape. Additionally Sigma is equivalent, or translates into, to the letter S in the latin alphabet (which english uses).

Because of the physical characteristics of the letter (being S-shaped), it is used to describe curves which exhibit the same properties. Specificially S-shaped curves.

# Logistic Function

## Overview

The logistic function is a function within the sigmoid family.

### Etymology

Verhulst originally published the logistic function in a paper written in his native French language. As such, he referred to the function as "logistique". The term logistic is an english translation of that term. Additionally, Verhulst did not provide an explanation, in his publication or otherwise, for why he chose the term. Below is a speculation that seems to be widely accepted; or atleast proposed as an infferential reasioning as to the word choice.

I believe we do get some insights an intuition from the arguments and discussion that Verhulst provides in his paper while arguing that the logistic curve better models population growth as compared to the logarithmic curve. I believe the chioce of terminology is ultimately made to characterize the curves according to their underlying philosophical beliefs.

Verhulst begins by discussing different models for population growth. He puts forward a number of equations which arethmetically appear exponential; for example one takes the form $p=k10^{lt}$ where p is the population, k and l are constants, and t represents time. During this section he discusses geometric growth rates, and he qualifies the population models as being logarithmic because they have similar characteristics to logarithmic functions; they start at the origin (denoting zero population at  the initial point in time) and have no maximum value. 

**Note**: I've seen some some confusion in various posts about this section and the implications of the use of the qualifier logarithmic. I believe many misunderstood or mistranslated the section and took it to mean that Verhulst was referring to these geometric and/or exponential forms as being THE logarithmic function, rather than just exhibiting properties of a logarithmic function.

He also criticizes these models because they do not model any forces that would put downward pressure on the growth rate or limit them with a horizontal asymtote.

### Quetelet (1835) - Constrained Population Growth

As [Bacaer (2011)](https://webpages.ciencias.ulisboa.pt/~mcgomes/aulas/dinpop/Mod13/Verhulst.pdf) notes:

> That same year 1835, his compatriot Adolphe Quetelet, a statistician and director

### Verhulst (1838) - Population Growth With Upper Limit

Verhulst [publishes](https://userpages.umbc.edu/~rostamia/math481/1838-Verhulst.pdf) the first argument for constrained population growth in a four page paper titled *Notice sur la loi que la population suit dans son accroissement*. This loosely translates to *Note on the law of population growth*. This was published in the french publication, *Correspondance mathématique et physique* journal, which translates to the Journal of Math and Physics. A full [english translation](https://www.neo-classical-physics.info/uploads/3/4/3/6/34363841/verhulst_-_law_of_population.pdf) is provided by Delphenich.

As [Bacaer (2011)](https://webpages.ciencias.ulisboa.pt/~mcgomes/aulas/dinpop/Mod13/Verhulst.pdf) provides a translated excerpt:

> We know that the famous Malthus showed the principle that the human population tends to
grow in a geometric progression so as to double after a certain period of time, for example
every twenty five years. This proposition is beyond dispute if abstraction is made of the
increasing difficulty to find food [. . . ]
The virtual increase of the population is therefore limited by the size and the fertility of
the country. As a result the population gets closer and closer to a steady state.

[Bacaer (2011)](https://webpages.ciencias.ulisboa.pt/~mcgomes/aulas/dinpop/Mod13/Verhulst.pdf) speculates the Verhulst did not completely agree with Quetelet; specifically taking issue with the mechanical nature of the analogy. This is not directly mentioned in Virhulst's first publication.

Reading through the translation by [Delphenich](https://www.neo-classical-physics.info/uploads/3/4/3/6/34363841/verhulst_-_law_of_population.pdf) we see:

> One can make various hypotheses in the resistance or the sum of the obstacles that are
opposed to the indefinite development of the population. Quetelet supposed that it was proportional to the square of the rate at which the population tends to grow.
>
> This equates the motion of the population to that of a moving body that falls while
traversing a resistant medium. The results of that comparison agree in a satisfactory
manner with the ones that are given by statistics and with the ones that I obtained from
own formulas when I assumed that there is an indefinitely-increasing density in the layers
of the medium being traversed.

We see Verhulst reiterate that there must be an upper limit to growth (Delphenich):

> The increase in the population necessarily has a limit that is defined by the ground
area that is available for the housing of the population... All of the formulas by
which one attempts to represent the law of population must then satisfy the condition that
they admit a maximum that is attained only after an infinitely-extended epoch.


Verhulst then gives an equation the widely cited differential equation

$$ \frac{dp}{dt} = mp - \phi(p) $$

Where $p$ is population, $m$ is an arbitrary constant (interpreted as a growth rate), and $\phi$ is an unknown function of $p$  (interpreted as the downard pressure or decay rate).

He notes that 

> If the population grows in a progression that is more than geometric then the term
−ϕ(p) will become +ϕ(p).

Verhulst then gives several example functions for $\phi$ such as: $\phi(p) = np^2$, $\phi(p) = np^3$, $\phi(p) = np^4$, $\phi(p) = n \ log \ p$ where $n$ is an arbitrary constant.

He notes that they will produce similar values but different upper limits.

He then compares actual populations with theoretical populations calculated using the proposed functions and notes that there is a similarity and shows that the error term is not growing. He does not specifically state which of these example functions he uses.

**Note**: Verhulst does not write the logistic equation as we know it today in this paper despite people citing it as the source. Below I will explain algebraically how to derive today's form; but this is not part of verhulsts paper.

If we expand the differential equation with $\phi(p) = np^2$ we will see 

$$ \frac{dp}{dt} = mp - np^2$$

And if we consider the arbitrary constant $n$ to be composed of additional arbitrary constants such that $n=\frac{m}{k}$ we see the classic form forthe derivative that is referenced today:

$$ \frac{dp}{dt} = mp - \frac{mp^2}{k}$$

$$ = mp(1 - \frac{p}{k})$$

**Note**: As we discuss later, this equation is mathematically equivalent to the equation produced by Reed and Pearl and thus the integral, i.e. the logistic function is the same function.

### Verhulst (1844) - A follow up

In 1844, Verhulst [published](http://www.med.mcgill.ca/epidemiology/hanley/anniversaries/ByTopic/Verhulst1844.pdf) *Recherches mathématiques sur la loi d'accroissement de la population*. Here he expands on his prior work from 1838.

He starts by talking about the geometric population, using a similar formula but different notation:

$$ p = \Pi 10 ^ {lt} $$

Where $p$ is again the population, $\Pi$ is the original population, and $l$ is an arbitrary constant and the growth rate.

In a lengthy discussion he tries to reverse-engineer changes into the equation; he looks at the first and second derivatives to understand if the conclusions being asserted by the equation's form are still valid. 

After reading through the paper, there was nothing major in terms of the development of the logistic function. The form published in his prior work was an early variant of the form recognized today. The versions of the equations published in this paper, and later papers, only manipulate constants involved in the algebra of fitting the curve to the data and do not materially impact the mathematical form produced by the original equation.

### Reed and Pearl (1920) - Logistic rediscovered
In 1920, Reed and Pearl [published](https://www.ncbi.nlm.nih.gov/pmc/articles/PMC1084522/pdf/pnas01915-0003.pdf) an article titled *On the rate of growth of the population of the united states since 1790 and its mathematical representation* in the journal named *Proceedings of the national achademy of sciences*.

With this publication the pair are credited with independently rediscovering the Logistic curve originally published by Verhulst and then popularizing it within the west.


They make a similar argument for the characteristics that a population model must satisfy:

> The following conditions should be fulfilled by any equation which is to
describe adequately the growth of population in an area of fixed limits.
> 1. Asymptotic to a line $y = k$ when $x=-\infty$ .
> 2. Asymptotic to a line $y=0$ when $x = +\infty$
> 3. A point of inflection at some point $x = \alpha$ and $y=\beta$.
> 4. Concave upwards to left of $x = \alpha$  and concave downward to right of $x = \alpha$ .
> 5. No horizontal slope except at $x=\pm \infty$
> 6. Values of y varying continuously from 0 to k as x varies from $-\infty$ to $+\infty$

They propose the following equation which they argue satisfy these requirements if $a$, $b$, and $c$ have positive values.

$$y = \frac{be^{ax}}{1+ce^{ax}}$$

They also note that this function is equivalent to this simpler version which they refer to as the final form:

$$y =\frac{b}{e^{-ax}+c}$$

$$y = \frac{K}{1+Ae^{-rt}}$$

$$= \frac{ayb(1-\frac{cy}{b})}{b} $$
$$= ay \big(1-\frac{cy}{b} \big)$$

### Constant Growth

If a population is growing at a constant rate, the derivative of that growth function is a constant

$$y = mx+b,$$
$$ \frac{dy}{dx}=m$$

### Geometric Growth

If a population grows at a constant proportion or percentage, i.e. a geometric rate, the population will grow by the same percentage each time period. The derivative of that growth function is not constant, but is instead proportional to the length of time. Each full time interval, the population is multiplied a given number of times coresponding to that time interval.

$$y = a(1 + r)^t,$$
$$ \frac{dy}{dt}= a(1 + r)^t \ ln(1+r) $$

And we can see this porportionality in the simplification of the derivative above.

$$ = y \ ln(1+r)$$ 

$$ = y \beta$$ 

Becasue of this, the geometric function is discrete as the time intervals are a finite and fixed length. If we think about continous or instantaneous growth, we are thinking about exponential growth.

### Exponential Growth

Exponential growth is modeled using the exponential function. This function usually takes the form: 

$$y = ae^{rt},$$
$$\frac{dy}{dt}=rae^{rt}=ry$$

Where $a$ is a starting point or initial population.

**Note**: I had come confusion as to why $e$ is generally selected as the base for the exponential function. To understand this, we need to work our way in reverse; we start thinking about the rate of change (i.e. what the derivative equals), we then integrate the derivative to get a solution for the equation. It just to happens that algebraically, the solution will always involve raising $e$ to some power. A proof can be found on this [stackexchange](https://math.stackexchange.com/questions/1350855/why-do-we-use-base-e-in-population-growth-questions).

As noted [here](https://www2.clarku.edu/faculty/djoyce/ma121/logistic.pdf), we can algebraically manipulate the exponential equation so that the base is something other than $e$. In order to do so, we must make a few adjustments:

$$y = Ab^{\frac{rt}{ln(b)}}$$

And we can simplify the equation with $s=\frac{r}{ln(b)}$ such that the equation becomes:

$$y = Ab^{st}$$

$$\frac{dy}{dt}=ln(b)sAb^{st} =ln(b)sy = ry$$

**Note**: We again see the proportionality in the equation above.

### Restricted Exponential Growth

You start with exponential growth

$$\frac{dy}{dt} = \beta y$$

Then add a downward pressure which is dependent on the population. 

We can do this in one of two ways:

First, by subtracting some variable from our exponential growth rate:

$$\frac{dy}{dt} = \beta y - \phi(y)$$

Or second we can multiply our original growth rate by some decreasing fucntion $g(y)$ so that the growth rate is again proportional to the population while having a downward pressure. 

$$\frac{dy}{dt} = \beta y \ g(y)$$

In either case, in order for the function to take on the requisite characteristics to become a logistic function, we would need to restrict these decay functions such that $\phi(y)= ny^2$ and $n=\frac{\beta}{k}$

$$\frac{dy}{dt} = \beta y - ny^2$$

$$ \frac{dy}{dt}=\beta y \big(1 - \frac{y}{k} \big)$$
