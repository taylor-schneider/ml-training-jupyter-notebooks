# Overview

The tool is provides a low-code drag & drop UI to build your customized LLM workflows. It is built on top of two major AI frameworks (LangChain and LLamaindex) and provides a rich set of third party integrations. The solution also allows users to define their own functionality by writing their own TypeScript snippets which can be embedded in the defined workflows.

The official github page can be found [here](https://github.com/FlowiseAI/Flowise) and the official documentation can be found [here](https://docs.flowiseai.com/).

# Solution Architecture

According to the [documentation](https://docs.flowiseai.com/getting-started), the FlowiseAI solution is stored in a single mono repository. This [repo](https://github.com/FlowiseAI/Flowise) contains the following three modules:

- server: Node backend to serve API logics
- ui: React frontend
- components: Integrations components




All of this functionality is defined as TypeScript and deployed by locally installing an NPM package and then running an included binary.

## Components

If we navigate to `packages/components/` within the repository we arrive at the directory containing all the code for the components. The documentation notes that the components consist of nodes and Credentials. Looking at the structure of the code, we can see subdirectories for both nodes and credentials.

### Nodes

After looking at the github page, we can see that nodes are defined in [packages/components/nodes](https://github.com/FlowiseAI/Flowise/tree/main/packages/components/nodes) directory. Looking through the repo, we can see that each node gets it's own directory. While the contents of the directory vary between nodes, we can see a consistency between nodes. Each is expected to have an image file (.svg extension) containing the icon for the UI, and typescript file (a variant of javascript) containing the code for the node. We see each node is defined as a class which extends the Node baseclass.

# Major Concepts

In the next section, I walk through some of the major concepts and functionality exposed through the UI. The official documentation can be found [here](https://docs.flowiseai.com/).

### Assistants & Agents

Generally speaking the concept of Assistant and Agents are used interchangeably. An Assistant or Agent is a wrapper around an LLM. The agent proxies requests from the user. In some cases, the agent will administer a prompt establishing the specific function and/or desired response format from the LLM.

Major LLM providers like OpenAI and IBM Watson offer Assistant APIs for their LLM offerings. But it is also possible for open source off-the-shelf agents to exist for the open-source LLM offerings.

The way to leverage an Assistant using FlowiseAI is through a Chatflow; specifically through the Nodes in the chatflow. FlowiseAI offers a number of Nodes providing Agent functionality which in turn wraps around an LLM. In the case of a vendor providing an assistant API, the FlowiseAI Agent is typically just a wrapper around the vendor's agent API.

This is the case when using OpenAI. When we create a new OpenAI Agent, FlowiseAI will perform the backend API calls to properly configure the Agent within OpenAI's platform. The FlowiseAI platform will then handle the logic of redirecting the user input provided to the chat bot to the OpenAI Assistant and then relaying the response back to the UI.

## Chatflows and Nodes

A Chatflow is FlowiseAI's abstraction for an AI workflow which is expected to provide some sort of chat bot functionality. The Chatflow is implimented as a Directed Asyclic Graph (DAG) which defines the flow of data between Nodes.

According to the [documentation](https://docs.flowiseai.com/integrations) **Nodes** are also referred to as **integrations** and serve as the foundational building block for Chatflows.

After looking at the source code, we can see that the Nodes provide the built-in integrations between the Chatflow and external services.

If we want to add additional functionality or integrations to a Chatflow, we can add a Node to the workflow by clicking the plus button on the top left corner of the charflow page.

### Framework Support

The Nodes provided by FlowiseAI are largely wrappers around functionality provided by two main typescript libraries: Langchain and Llamaindex. These tools have some overlap, but they are largely viewed as complementary within the industry and not as being redundant. For example, see this [stackoverflow discussion](https://stackoverflow.com/questions/76990736/differences-between-langchain-llamaindex).

The Nodes offered by the two technologies are not the same; some Nodes available from the Langchain menu are not available from the Llamaindex, and vice versa. Ironically, around this time, Facebook also released it's Llama model which added some confusion.

#### Langchain

Langchain is described as a framework for developing applications powered by large language models (LLMs). Through its functionality and integrations, LangChain aims to simplify every stage of an LLM's application lifecycle including development, productionalization, and deployment.

LangChain offers several smaller components such as langgraph and langserve. And is also part of a larger ecosystem which includes tools like LangSmith.



#### LlamaIndex (Formerly GPT Index)

In short, LLamaindex is a context augmentation framework for LLMs and is provided as a python pip package and a typescript npm package.

**Note**: Do not confuse the opensource LlamaIndex tool with the llamaIndex concept/product as defined by [lark](https://www.larksuite.com/en_us/topics/ai-glossary/llamaindex).

According to the official [homepage](https://www.llamaindex.ai/open-source), LlamaIndex is described as a simple, flexible data framework for connecting custom data sources to large language models. Looking at the [documentation](https://docs.llamaindex.ai/en/stable/), we see a more elaborate definition:

> LlamaIndex is a data framework for LLM-based applications which benefit from context augmentation. Such LLM systems have been termed as RAG systems, standing for "Retrieval-Augmented Generation". LlamaIndex provides the essential abstractions to more easily ingest, structure, and access private or domain-specific data in order to inject these safely and reliably into LLMs for more accurate text generation.

Apparently LlamaIndex was renamed from GPT Index around mid February 2023 due to naming conflicts and potential trademark violations with OpenAI's GPT Brand as noted in these posts: [ycombinator post](https://news.ycombinator.com/item?id=35973645), [ycombinator post](https://news.ycombinator.com/item?id=36642536), [ycombinator post](https://news.ycombinator.com/item?id=34928386), [twitter post](https://twitter.com/jerryjliu0/status/1650673187896844288).

### Agents

Generally speaking, an Agent is FlowiseAI's implementation of an Assistant. It is a wrapper around an LLM aiming to provider a chat bot-like user experience where the agent proxies requests from the user. In some cases, the agent will administer a prompt establishing the specific function and/or desired response format from the LLM.

FlowiseAI offers a number of Nodes providing Agent functionality which in turn wraps around an LLM. In many cases, the provider of the LLM may also provide an API for defining an Agent. In such cases, the FlowiseAI Agent is then typically just a wrapper around the vendor's agent API. 

This is the case when using OpenAI. When we create a new OpenAI Agent, FlowiseAI will perform the backend API calls to properly configure the Agent within OpenAI's platform.

**Note**: While the built-in Agents are useful, they are opinionated approaches to interacting with the LLM. In some cases is might be better to implement the logic and prompting for an agent from scratch using the lower level Prompt, Chain, and LLM Node Types. For example when prompt chaining.

### Cache

Generally speaking, Caching refers to the process of moving information from a systems memory to another location. Traditionally this would be a transition to disk, however, with the creation of caching servers, this could be the memory of another networked system.

The collection of Cache Nodes provide the ability to integrat a Chatflow with various services allowing for LLM responses and Embeddings to be cached. These services include Redis, Upstash, and Momento.

### Chains

While Chains have a framework specific implementation, they can be generally defined as being a sequential set of operations that a framework is used to perform, which may be interdependent.

Confusingly, while a chain conceptually represents a collection of actions, the Chain Nodes in Flowise represent a single action. Thus if we wanted to have multiple actions chained together, we would need to define multiple Chain Nodes and connect them accordingly. 

A great video for performing prompt chaining through FlowiseAI can be found [here](https://www.youtube.com/watch?v=yLfiNnK4NOM). In this example, a chef assistant creates a recipe which uses a user provided ingredient and then the recipe is critiqued by a food critic assistant.

Currently FlowiseAI only provides Chain Nodes for the LangChain framework.

### Chat Models

These Nodes serve as wrappers around Chat-style APIs and/or LLMs. According to the [documentation](https://docs.flowiseai.com/integrations/langchain/chat-models), the following integrations are provided:

* Azure ChatOpenAI - Allows connecting to an OpenAI provided model deployed through Azure
* ChatGoogleGenerativeAI - Allows connecting to a Google provided model
* ChatLocalAI - Allows connecting to a self hosted LocalAI API instance
* ChatOpenAI - Allows connecting to an OpenAI provided model directly
* GoogleVertexAI - Allows connecting to a Google VertexAI provided model
* ChatMistralAI - Allows connecting to a MinstralAI provided model

### Document Loaders

According to the [LangChain documentation](https://python.langchain.com/docs/modules/data_connection/document_loaders/), a Document Loader is an abstraction used to load data into a "documents" format; i.e. json representations of the loaded information, and then store that representation in memory.

From looking at the [documentation](https://docs.flowiseai.com/integrations/langchain/document-loaders), I believe FlowiseAI provides abstractions around the out of the box LangChain integrations as well as some which are not based on out-of-the-box functionality.

### Embeddings
This group of Nodes provides integrations with various embeddings technologies.

### LLMs
This group of Nodes provides integrations with various LLMs.

### Memory
This group of Nodes provides the ability to store chat history and other information to storage mediums and then access that information as part of the Chatflow. The mediums include Redis, DynamoDB, Motorhead, and more.

### Moderation
These Nodes help with prompt and output moderation and are used to ensure that no blacklisted topics or content are exchanged with the user of the Chatflow.

### Output Parsers
The Nodes are used to perform transformations on the LLM outputs.

### Prompts
These Nodes allow the user to specify a prompt for a given LLM interaction. They can be used to custom define chat assistants.

### Record Managers

A Record Manager is an abstraction that keeps track of writes to a vector store. They typically store the data and metadata of a record and the relation with it's vector. The purpose is to prevent duplicate conversions and writes of documents that have already been indexed and to provide faster retrieval times for previously converted documents.

For more information see the [LangChain documentation](https://python.langchain.com/docs/modules/data_connection/indexing/) or the [FlowiseAI documentation](https://docs.flowiseai.com/integrations/langchain/record-managers).

### Retrievers
From looking at the descriptions of the Nodes in this section, it appears that they all have to do with retrieving documents from a set of documents based on semantic similarity. As is common in RAG implementations, a conceivable pattern would be to find documents related to a query so that the LLM's context can be augmented with the content of the relevant documents.

There appear to be several Nodes being offered which wrap around and integrate with various providers.

### Text Splitters

Because LLMs have a finite context window, doing context augmentation effectively often requires text to be split up into smaller chunks so that it can be selectively consumed if and when it becomes relevant to an interaction.

These Nodes provide the basic functionality related to splitting and combining split chunks of text.

For more information please see this [documentation](https://docs.flowiseai.com/integrations/langchain/text-splitters).

### Vector Stores

These Nodes provide integrations with propular Vector Stores allowing the Chatflow to access the vectors stored in the Vector Store.

### Tools

According to the [documentation](https://docs.flowiseai.com/integrations/langchain/tools):

> Tools are functions that agents can use to interact with the world. These tools can be generic utilities (e.g. search), other chains, or even other agents.

From playing with the solution and reading through the documentation, I believe the tools are implemented in several ways. There are built-in Tools and then there are user defined tools.

#### Built-In Tools (Chatflow Nodes)

The [official documentation](https://docs.flowiseai.com/integrations/langchain/tools) about tools describes them as:

> Tools are functions that agents can use to interact with the world. These tools can be generic utilities (e.g. search), other chains, or even other agents.


and links to a subdirectory within the nodes directory. This [tools directory](https://github.com/FlowiseAI/Flowise/tree/main/packages/components/nodes/tools) contains all the definitions for the tools available to the FlowiseAI system.

Similar to the nodes, each tool has it's own sub directory. The tool directories are expected to contain image SVG files and typescript files which contain classes that impliment the INode interface.

The tools defined in this directory are visible when we look to add a node to a chatflow and expand the tools menu.

After looking at the source code, I believe some of these are wrappers around functionality provided by the [lanchainjs library](https://github.com/langchain-ai/langchainjs) or other libraries. 

For example, the Calculator Tool Node has the following reference

```javascript
import { Calculator } from 'langchain/tools/calculator'
```

If we track this back the the langchainjs github page, we can find the definition of the calculator class [here](https://github.com/langchain-ai/langchainjs/blob/63a0ec27f9136ae40dfcf1d0eff6683dd5eaa5a2/libs/langchain-community/src/tools/calculator.ts). We see that this javascript class impliments a _call() fucntion which accepts an input string and attempts to evaluate it using a parser. If the expression is a valid math expression, the function will return the result.

#### Custom User Defined Tools

The same [documentation](https://docs.flowiseai.com/integrations/langchain/tools/custom-tool) then does on to describe how to create custom tools. Here The Tools are created by using the main Tools option from the main menu. Once added, these tools can be used in a Chatflow by adding a "Custom Tool" node and then selecting the custom Tool from a dropdown.

This allows us to define our own javascript functions which can then be included in our Chatflows.

## Marketplace & Templates

According to the [documentation](https://www.restack.io/docs/flowise-ai-knowledge-flowise-ai-marketplace), The Marketplace is designed as a centralized repository which allows users to publish and consume various templates which can be used when creating Chatflows. These templates contain a reusable skeleton definition for a chatflow which users can then configure for their usecase. 

The Marketplace is intended to provide the following features:

> - Integration with Third-Party Platforms: Flowise can be integrated with platforms like Zapier, enhancing automation capabilities.
> - Diverse Use Cases: From web scraping Q&A to webhook tools, the marketplace supports a variety of applications.
> - Custom Chat Models: Users can drag and drop pre-built nodes like Azure ChatOpenAI and configure them with ease.
> - Deployment Flexibility: Flowise supports deployment across multiple cloud services, including AWS, Azure, and GCP.
> - Streaming: Flowise supports streaming responses to front-end applications.
> - LocalAI Integration: Users can set up LocalAI embeddings for advanced text analysis.

Within the FlowiseAI application, the marketplace menu serves as the connection to the centralized Marketplace. The interface to empowers users to build LLM apps using a visual drag-and-drop interface.

# Installation

Currently, the latest version of Flowwise (1.6.4) is packaged as a node module and can be installed using the npm utility. The github page specifies that NodeJS >= 18.15.0 must be used.

Below we can see the commands I used to install the solution into a docker image.

```
(base) [root@fedora flowwise-test]# cat Dockerfile
FROM python:3.11.3-bullseye

# Install debugging utilities

RUN apt-get -y update
RUN apt-get -y install iproute2 net-tools traceroute vim

# Install node and required packages

RUN apt-get -y update
RUN curl -sL https://deb.nodesource.com/setup_18.x | bash -
RUN apt-get install -y nodejs
RUN node --version
RUN npm --version
(base) [root@fedora flowwise-test]#  docker build -t flowwise-test .
```

And here are the commands I used to run the container, then execute the command to install inside the container, and finally run an instance of FlowiseAI.

```
(base) [root@fedora flowwise-test]#  docker run -ti --net=host flowwise-test /bin/bash
root@fedora:/# node --version
v18.20.2
root@fedora:/# npm install -g flowise
root@fedora:/# npx flowise start --FLOWISE_USERNAME=user --FLOWISE_PASSWORD=1234
```
