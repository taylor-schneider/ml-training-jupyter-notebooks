# Overview

AutoGen Studio is an AI Studio written in python and ReactJS and published by Microsoft. The solution is built on top of the [AutoGen](https://github.com/microsoft/autogen/tree/main/autogen) framework which is also written in python and published By microsoft. AutoGen is billed as a framework for orchestrating GenAI Agents.

# Major Concepts

## AutoGen Framework

AutoGen is billed as a framework for orchestrating GenAI Agents. This framework can be viewed as a competitor to the popular LangChain framework.

## UI Design

The AutoGen Studio UI is organized into 3 high level sections - Build, Playground, and Gallery.

### Build Section

As the name suggests, the build section is concerned with building AI workflows. As such, the section allows you to declare and configure the basic building blocks of the platform including: Skills, Models, Agents, and Workflows. Together these components allow the user to interact with live systems in the Playground Section

#### Skills

According to the [documentation](https://microsoft.github.io/autogen/blog/2023/12/01/AutoGenStudio/):

> Skills are functions (e.g., Python functions) that describe how to solve a task. In general, a good skill has a descriptive name (e.g. generate_images), extensive docstrings and good defaults (e.g., writing out files to disk for persistence and reuse). You can add new skills to AutoGen Studio via the provided UI. At inference time, these skills are made available to the assistant agent as they address your tasks.

#### Models

From poking around through the UI and parsing through the documentation we can infer that a Model is a wrapper around an LLM which allows a user to interface with and interact with the LLM. While AutoGen standardizes on the openai model api format, and you can use any api server that offers an openai compliant endpoint.

#### Agents

Parsing through the documentation, the Agents are abstractions for elements in a workflow. An agent might wrap around an LLM and allow text to be input and output, an agent might provide the user with an interface to interact with other agents, or orchestrate communications between multiple agents. 

According to the [AutoGen Studio documentation](https://microsoft.github.io/autogen/blog/2023/12/01/AutoGenStudio/):

> This provides an interface to declaratively specify properties for an AutoGen agent.

According to the [AutoGen documentation](https://microsoft.github.io/autogen/docs/Use-Cases/agent_chat/):

> AutoGen abstracts and implements conversable agents designed to solve tasks through inter-agent conversations.
>
> The figure below shows the built-in agents in AutoGen.
>
> <center><img src="images/autogen_agents.png"></center>
>
> We have designed a generic ConversableAgent class for Agents that are capable of conversing with each other through the exchange of messages to jointly finish a task. An agent can communicate with other agents and perform actions. Different agents can differ in what actions they perform after receiving messages. Two representative subclasses are AssistantAgent and UserProxyAgent
>
> - The AssistantAgent is designed to act as an AI assistant, using LLMs by default but not requiring human input or code execution. It could write Python code (in a Python coding block) for a user to execute when a message (typically a description of a task that needs to be solved) is received. Under the hood, the Python code is written by LLM (e.g., GPT-4). It can also receive the execution results and suggest corrections or bug fixes. Its behavior can be altered by passing a new system message. The LLM inference configuration can be configured via [llm_config].
>
> - The UserProxyAgent is conceptually a proxy agent for humans, soliciting human input as the agent's reply at each interaction turn by default and also having the capability to execute code and call functions or tools. The UserProxyAgent triggers code execution automatically when it detects an executable code block in the received message and no human user input is provided. Code execution can be disabled by setting the code_execution_config parameter to False. LLM-based response is disabled by default. It can be enabled by setting llm_config to a dict corresponding to the inference configuration. When llm_config is set as a dictionary, UserProxyAgent can generate replies using an LLM when code execution is not performed.
>
> The auto-reply capability of ConversableAgent allows for more autonomous multi-agent communication while retaining the possibility of human intervention. One can also easily extend it by registering reply functions with the register_reply() method.

#### Workflows

According to the [documentation](https://microsoft.github.io/autogen/blog/2023/12/01/AutoGenStudio/), a workflow is a collection of agents:

> An agent workflow is a specification of a set of agents that can work together to accomplish a task. The simplest version of this is a setup with two agents – a user proxy agent (that represents a user i.e. it compiles code and prints result) and an assistant that can address task requests (e.g., generating plans, writing code, evaluating responses, proposing error recovery steps, etc.). A more complex flow could be a group chat where even more agents work towards a solution.

### Playground Section

The Playground section allows users to instantiate and interact with the workflows defined in the Build section. The main elements a user interacts with in the playground are the workflow and the sessions.

#### Session

According to the [documentation](https://microsoft.github.io/autogen/blog/2023/12/01/AutoGenStudio/):

> A session refers to a period of continuous interaction or engagement with an agent workflow, typically characterized by a sequence of activities or operations aimed at achieving specific objectives. It includes the agent workflow configuration, the interactions between the user and the agents. A session can be “published” to a “gallery”.

The UI allows us to logically partition our interactions with a Workflow into separate Sessions.

### Gallery Section

This section is focused on sharing and reusing artifacts (e.g., workflow configurations, sessions, etc.).

AutoGen Studio comes with 3 example skills: fetch_profile, find_papers, generate_images.

# Installation

AutoGen Studio is written in python and distributed as a python module. It can be installed with pip as follows:

```
(autogen) [root@fedora ~]# pip install autogenstudio
```

Currently v0.0.56 is the latest available release. It requires python >= 3.9.

**Note**: While the solution is distributed as a pip package, looking at the [documentation](https://github.com/microsoft/autogen/tree/main/samples/apps/autogen-studio) we can see it also leverages ReactJS and Node.js for the front end.

Once installed the solution can be run with the following command:

```
(autogen) [root@fedora ~]# autogenstudio ui --host 0.0.0.0 --port 8081
```

**Note**: Here we are instructing which IP and port the service should run on.
