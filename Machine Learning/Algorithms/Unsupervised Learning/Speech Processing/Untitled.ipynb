# Overview

In general, speech or audio processing is a field of computer scince concerned with analyzing and interpreting auditory language data. Typically this data is presented as either a file containing or stream of an audio recording. The main objective of the speech processing process is to extract signals and information from the raw recording data.

More recently, with the advancement of machine learning and deep neural networks (DNNs), the field has seen many novel approaches to speech processing. 

# Major Topics

## Major Audio/Speech Processing Tasks
When processing audio data contaiing speech, there are a number of terms which enter the conversation and can often be conflated. As we will see, this is often the case bcause many of these techniques are intterrelated or performed in parallel.

- Transcription - Determine the words and punctuation associated with the associated audio data.
- Language Detection - Detemine what language(s) are being spoken.
- Speaker Identification - Destinguishing who is speaking (from those identified in the recording) (e.g. is this speaker 1 or speaker 2), or destinguishing the personal identity of a particular speaker (e.g. is jim speaking)
- Speaker Verification (voice fingerprinting) - Confirming a particular individual is speaking
- Spech Diarization - Partitioning the audio file into segments which are mapped to who is speaking
- Sentiment Analysis - Analyzing the mood, emotion, or opinion of the speaker

A helpful article can be found [here](https://www.assemblyai.com/blog/speaker-diarization-vs-recognition)

As we can see, speeh processing breaks dow into many subtasks. And it can be confusing to understnad what a speech processing library or pipeline is doing. Additionally, one of the most widely used terms is:

- Speech Recognition - A broad term used interchangibly with many of the terms above. Sometimes aslo called Automatic Speech Recognition (ASR)

As we will see speech processing or speech recognition libraries and toolkits usually perform more than one of the functions listed above. As such, these libraries end up being ollctions of machine learning models which work, and sometimes train, jointly or having interdependencies.



## Waveform & Sound Waves

There are many conceptual or mathematical models used to explain the concept and phenomenon of sound. It is generally explained a brain's perception of a vibration of a medium, or the energy passing through a medium, or the contraction and expansion of molecules within a medum, as it reaches our ear.

The commonality between these differing explanations is that sound (i.e. the energy associated with the phenomenon) is understood to propogate or travel as a waive through the medium. Through many experiments and through our reproduction of sound, we can clearly see that it is a wave. For example, stereo speakers and bone conduction headphones both vibrate in order to produce sound.

### Transverse and Longitudinal Waves
There are two types of waves, or wave forms, used to describe the movement of energy through a medium or system: transverse and longitudinal.

With a longitudinal wave, we can see the particles in the medium move parallel with the direction of the energy propogation.

>Sound is transmitted through gases, plasma, and liquids as longitudinal waves, also called compression waves...Through solids, however, it can be transmitted as both longitudinal waves and transverse waves.

In some cases, sound may be visually represented as a transverse wave rather than a longitudinal wave. From what I understand this is for simplicity and convenience; with the peaks representing compressions and troughs representing rarefactions in the actual longitudinal wave.

We will see with WAV files that this is also done because ofhow the recording or playback hardware is designed.

### WAV Files

WAV files are often used for speech processing because they store the "raw" and often uncompressed audio data which is considered to be the truest and richest representation of the original signal.

According to wikipedia:

> Waveform Audio File Format (WAVE, or WAV due to its filename extension) is an audio file format standard for storing an audio bitstream on personal computers. The format was developed and published for the first time in 1991 by IBM and Microsoft. It is the main format used on Microsoft Windows systems for uncompressed audio. The usual bitstream encoding is the linear pulse-code modulation (LPCM) format.
>
> Though a WAV file can contain compressed audio, the most common WAV audio format is uncompressed audio in the linear pulse-code modulation (LPCM) format. LPCM is also the standard audio coding format for audio CDs, which store two-channel LPCM audio sampled at 44.1 kHz with 16 bits per sample. Since LPCM is uncompressed and retains all of the samples of an audio track, professional users or audio experts may use the WAV format with LPCM audio for maximum audio quality.

A good explanation of what's in a WAV file can be found in this [stackoverflow article](https://stackoverflow.com/questions/71022710/how-is-wav-data-stored-in-principle) and this [reddit post](https://www.reddit.com/r/DSP/comments/ipnn9a/trying_to_understand_wav_files/).

Within the WAV file there is a header, specifying the variations (sampling rate, bit depth, number of channels etc), and there is the raw data. From what I understand, the raw data is a transverse wave due to how computers record or generate sounds. When recording sounds, the data is simply the strenght of the signal the microphone is measuring the sound wave or the voltage induced in a speaker producing the wave. It is referred to as the amplitude because multiple measurements are recorded over time which allows us to consider the voltage as a wave and thus the measurement as the amplitude.

## Channel / Session

There are many definitions for a channel. It is sometimes considered a path through which sound can travel to or from a specific location. For example, a microphone produces one channel of audio, and a speaker can accept one channel. In the case of digital audio, a channel is a discrete stream of audio data encoded into your source material like a Blu-Ray or CD. Stereo data is a term used to describe dual channel audio data which has parallel streams of audio information which are sent to the lft and right speaker or headphone. More recent 5 and 7 channel audio data is uses to send sounds through a larger array of speakers which allows for more sophistocated wideness or panning effects between the speakers in the sound system.

We can think of each channel as capturing or carrying a discrete wave form that was recorded by the audio capture device. The device may be operating accross multiple channels in parallel (eg. stereo) and thus the original sound wave migh be expressed as two discrete waveforms expressed through the channel.

In speech recognition the channels are processed to extract speaker or speach information. Depending on the approach the channels may be processed in parallel or one at a time.

## Utterance

The term utterance refers to a continuous piece of speech "uttered" or spoken by one person which is preceeded and followed by silence from the uttering person.

The coining of the term is commonly attributed to Russian philosopher Mikhail Bakhtin who extensively explored the concept of the utterance within his theories of language and discourse, particularly in his work "The Problem of Speech Genres."

In many speach recognition tasks, utterances are used as the base unit of measurment or concern. For example, we see utterances converted into vectors or embeddings as part of a speker identification or speach recognition.

For more information see this [wikipedia article](https://en.wikipedia.org/wiki/Utterance).

## Speaker Embeddings

Like other embeddings (e.g. [word embeddings](../Large%20Langugage%20Models/Word%20Embeddings.ipynb)), speaker embeddings are n-dimensional vectors that are learned from a corpus of training data. Similarely, each of the n-dimensions corresponds to some meaningul measurement or interpretation of the data. In the case of word embeddings, the dimensions are typically thought of as representing the strength of the relationship between a given word and the other words/meaning represented by each dimension. In the case of speaker embeddings, the embedding captures the relationship between the speaker and various speech characteristics represented by each dimension (e.g. tone, pitch, accent). 

The dimensions are learned, in an unsupervised fashion, by training a neural network to optimize acording to a similarity score. The idea being that embeddings from the same speaker should be most similar and thus should have an optimal similarity score.

Models can be trained to produce speaker embeddings given an audio segment for a speaker. They are essentially trained on numerous examples of labled speech data (i.e. waveforms we know were generated by a given speaker); For each iteration they are asked to generate an embedding for the wafeform and are repeatedly optimizied to identify important features of the waveform such that the rendered embeddings relaibly distinguish between individual speakers. I.e. the similarity scores between the embeddings generated for a particular speaker's waveform are better than those between two different speakers.

> Embeddings are high-dimensional numerical representations of data, where each embedding encodes meaningful patterns or characteristics about the data. In the context of speaker identification, embeddings represent unique voice features of a speaker, capturing nuances like pitch, tone, and speech patterns. These embeddings are vectors, and speakers with similar voice characteristics will have similar embeddings, while those with different characteristics will have more distinct embeddings.

# History

Deep learning and joint optimization

## Dehak et al. (2010) - i-vectors

In their 2010 [paper](https://sls.csail.mit.edu/publications/2010/Dehak_IEEE_Transactions.pdf) titled *Front-end factor analysis for speaker verification*, Dehak et al. provide a novel low dimensional representation of the audio information required for speaker identification.

> This paper presents an extension of our previous
work which proposes a new speaker representation for speaker
verification. In this modeling, a new low dimensional speaker- and
channel-dependent space is defined using a simple factor analysis.


The paper acknowledges that Joint Factor Analysis (JFA) had recently been used for speaker identification tasks and can be leveraged to model the inter-speaker variability and to compensate for channel/session variability in the context of Gaussian Mixture Models (GMM). GMMs are models which predict which cluster a given data point belong to; this is used as part of speaker recognition tasks etc.

Over several papers, Dehak et al. perform several experiments to show that the channl factors estimated by JFA contain speaker information. In this paper, the factor analysis is uses as a feature extractor. The factors then represent a lower dimensional space.


> In this new space, a given
speech utterance is represented by a new vector named total
factors (we also refer to this vector as “i-vector” in this paper).
The channel compensation in this new approach is carried
out in low-dimensional space, the total variability space, as
opposed to the high-dimensional GMM supervector space
for classical JFA.

## Li et al.(2017) - Deep Speaker

In their 2017 [paper](https://arxiv.org/abs/1705.02304) titled *Deep Speaker: an End-to-End Neural Speaker Embedding System*, Li et al. propose a novel method for training speaker embeddings using deep learning models.

The abstract states:

> We present Deep Speaker, a neural speaker embedding system that
maps utterances to a hypersphere where speaker similarity is measured by cosine similarity. The embeddings generated by Deep
Speaker can be used for many tasks, including speaker identification,
verification, and clustering. We experiment with ResCNN and GRU
architectures to extract the acoustic features, then mean pool to produce utterance-level speaker embeddings, and train using triplet loss
based on cosine similarity. Experiments on three distinct datasets
suggest that Deep Speaker outperforms a DNN-based i-vector baseline.

## Snyder et al. (2018) - X-vectors

In April, 2018, Synder et al. [published](https://www.danielpovey.com/files/2018_icassp_xvectors.pdf) *X-Vectors: Robust DNN Embeddings for Speaker Recognition*. The paper proposes adding noise and reverberations to existing labled training data to expand the amount of training data for training embeddings to discriminate between speakers. 

The embedding models being trained are deep neural networks (DNNs) which map variable length utterances to fixed dimensional embeddings. The paper distinguishes the rendered x-vectors from the previous generation of embeddings called i-vectors.

According to the paper:

> Most speaker recognition systems are based on i-vectors [2].
The standard approach consists of a universal background model
(UBM), and a large projection matrix T that are learned in an unsupervised way to maximize the data likelihood. The projection maps
high-dimensional statistics from the UBM into a low-dimensional
representation, known as an i-vector. A probabilistic linear discriminant analysis (PLDA) [3] classifier is used to compare i-vectors, and
enable same-or-different speaker decisions [4, 5, 6].

> The x-vector model, for instance, is based on a Time Delay Neural
Network (TDNN) and is now a fundamental component of the
state-of-the-art diarization systems [2].

## Ravanelli & Bengio (2018) - SincNet

In July or 2018, Ravanelli & Bengio [published](https://arxiv.org/abs/1808.00158) *Speaker Recognition from Raw Waveform with SincNet*.

At the time of writing, the usage of deep learning, spcifically Convolutional Neural Networks (CNNs) for speaker recognition tasks hs been growing as results showed potential as a viable alternative to i-vectors. An important distinction is that rather than employing standard hand-crafted features, the CNNs dynamically learn low-level speech representations from waveforms.

The paper proposes a novel CNN architecture which gets it's name from the network's reliance on parametrized sinc functions. The nodes and layers of the network act as audio filters which be tuned to identify and transform audio features. The method offers a very compact and efficient way to derive a customized filter bank specifically tuned for the desired application.

According to the authors:

> Our experiments, conducted on both speaker identification and speaker verification tasks, show that the proposed architecture converges faster and performs better than a standard CNN on raw waveforms.

## Dawalatabad et al. (2021) - ECAPA-TDNN model applied to speaker diarization

In April 2021, the paper *ECAPA-TDNN Embeddings for Speaker Diarization* was [published](https://arxiv.org/abs/2104.01466).

The authors explain:

> Speaker diarization answers the question of “who spoke when?”
in a given conversation [1, 2]. Diarization is used in many conversational AI systems and applied in various domains such

The abstract states:

> Learning robust speaker embeddings is a crucial step in speaker diarization. Deep neural networks can accurately capture speaker discriminative characteristics and popular deep embeddings such as x-vectors are nowadays a fundamental component of modern diarization systems. Moreover, we improved its robustness with a powerful augmentation scheme that concatenates several contaminated versions of the same signal within the same training batch. ## Stan (2023) - Analysis of Residual Information in Spearker Embeddings

In February 2023, Adriana Stan [published](https://arxiv.org/abs/2302.02742) *Residual Information in Deep Speaker Embedding Architectures*.

The paper abstract starts by explaining what speaker embeddings are

> Speaker embeddings represent a means to extract representative vectorial representations from a speech signal such that the representation pertains to the speaker identity alone. The embeddings are commonly used to classify and discriminate between different speakers.

It then goes on to explain the shortcommings of current embedding techniques:

> There is no objective measure to evaluate the ability of a speaker embedding to disentangle the speaker identity from the other speech characteristics. This means that the embeddings are far from ideal, highly dependent on the training corpus and still include a degree of residual information pertaining to factors such as linguistic content, recording conditions or speaking style of the utterance.

The paper analyzes several of the current deep learning approaches and architectures to see how well they are able to disentangle speaker identity from the speash signal.

The paper presents the results which show that the approaches retain residual information:

> The results show that the discriminative power of the analyzed embeddings is very high, yet across all the analyzed architectures, residual information is still present in the representations in the form of a high correlation to the recording conditions, linguistic contents and utterance duration.

## Jakubec et al. (2024) - Deep speaker embedding Review

In January 2024, Jakubec et al. [published](https://www.sciencedirect.com/science/article/abs/pii/S0952197623014161) *Deep speaker embeddings for Speaker Verification: Review and experimental comparison*.

The paper starts our by giving a nice introduction into the world of speaker recognition. It first dscribes the manual speaker recognition process that a human performs:

> Humans can extract speaker-specific voice features (e.g., pitch, timbre, prosody), crafting latent speaker identity representations for recognition despite challenging acoustic conditions. In speaker identification, invariant features persist consistently, distinguishing individual voices even amidst variations such as emotions or health conditions (Hansen and Hasan, 2015).

It then goes on to discuss how machines perform this same task:

> Automatic speaker recognition is traditionally divided into two tasks: identification and verification. Speaker Identification (SI) is a task in which a person is recognized among multiple enrolled speakers. SI is categorized as a closed-set identification problem in cases where the test utterance is produced by one of the speakers already included in the enrolled set. When the identification involves new speakers not included in the system’s training data, it is categorized as an open-set problem. Speaker Verification (SV) involves assessing whether a person’s claimed identity is valid, leading to either acceptance or rejection based on the score. As noted in Bai and Zhang (2021), SV can be regarded as a specific instance of open-set speaker identification.

The stated purpose of the paper is:

> to review the recent progress in speaker embedding development and to perform an experimental benchmark experimental comparison among the state-of-the-art deep speaker representations for a Speaker Verification (SV) task.
