# Major Concepts

### Convolutional Neural Networks (CNNs)

CNNs are not fully connected networks. Instead each neuron in the hidden and output layers is only connected to a subset of the layers in the preceeding layer. Additionally, this network uses a kernel to perform a convolution on the subset of neurons it is connected to. This is a fancy way of describing the activation function for the paritlaly connected network.

## Recurrent Neural Networks (RNNs)

An RNN is a neural network which processes data sequentially, unlike the simpler feed forward network which processes all the data in parallel.

Additionally, unlike a feed forward network, An RNN is bidirectional. Meaning that it is able to pass the output from a neuron back to a neuron in the same level rather than being required to only pass information to the next layer.

More complex networks are possible where a given node has access to more than just the immediately preceeding hidden state. Notable examples include Elman and Jordan networks.

## Intra vs Inter-attention (cross-attention)

Intra and inter-attention are basically opposites with respect to their consideration of sequences.

According to [Khoshsirat et. al (2023)](https://arxiv.org/abs/2309.11593),

>  Self-attention
(also known as intra-attention) is a type of attention that
quantifies the interdependence within the elements of a single input. At the same time, cross-attention (also known
as inter-attention) finds the interdependence across two or
more inputs [40, 18, 17].

Intra and inter-attention commonly appear together. *Vaswani et al (2017)* is one of the first examples I could find of this. Although in this paper, they use the term **cross-attention** rather than inter-attention.

As [Khoshsirat et. al (2023)](https://arxiv.org/abs/2309.11593) notes this co-appearence is not limited to seq2seq mapping, 

> Usually, cross-attention methods are used for multimodal inputs [43]. Cross-attention
models first process individual modalities using modalityspecific encoders, then the encoded features are fed into
cross-attention modules.

We can see examples of inter-attention in [Yang et. al. (2018)](https://arxiv.org/abs/1810.05866) and [Chen et. al. (2017)](https://arxiv.org/abs/1704.00763). In these papers the application is for computer vision. We see that the inter attention module is considering information from outside the input sequence.

While inter-attention does allow access to external information, it's not necessarily the case that the scope of the access is global. In other words, the mechanism may have acces to some, but typically not all external sequences.

# Etymology

Based on my review of the papers I could find, conversation with knowledgable colleagues, are a review of related blogs like [Hines - A Lit Review of Attentional Mechanisms](http://keeganhin.es/blog/attn.html) I was able to derive an understanding of what the term means.

It was very difficult to pinpoint the origin of the term self-attention. The term itself did not appear until relatively recenty however it is used in reference to papers which do not use the term. The earliest paper I could find using the term self-attention was Lin et. al. (2017). This paper references papers which instead use the term intra-attention.

The first question I had was "where did this term first appear?". *Lin et. al (2017)* seem to be the first to reference intra-attention. The next question was "did they invent self-attention or intra-attention?". It seems they did not. They did not provide a definition and they reference some previous works qualified as relating to self-attention we see in the [History section](Self-Attention.ipynb#History) of this notebook. Unfortunately these earlier works also do not define the term.

Historically, as attention was being developed, it was done so in the context of sequence to sequence mapping. As was the convention, in this scenario, attention is defined between the input sequence and the output sequence (for example with Bahdanau et. al. (2015)). In these circomstances, the attention scores are thus measuring the importance of the input tokens to a given output token. The scores are calculated using a function that takes the input and output sequences as inputs. This is generally referred to as intra-attention or cross attention.

In alternate circomstances however, we may not be doing sequence to sequence mapping, or we might be investigating alternate approaches where we use the same sequence twice in our attention calculation. Thus the attention score is measuring the importance or relationships between a given token and every other token in the sequence, including itself. Many cite this as being the key disctinction and reason for the term self being added as a prefix to the attention mechanism.



We see this is consistent with Viswani et. al. (2017) as they acknowledge in that paper:

> Self-attention, sometimes called intra-attention is an attention mechanism relating different positions
of a single sequence in order to compute a representation of the sequence. Self-attention has been
used successfully in a variety of tasks including reading comprehension, abstractive summarization,
textual entailment and learning task-independent sentence representations [4, 22, 23, 19].

We see that these inferences are consistent with *Lin et. al (2017)* who use the term "intra-attention", as well as with other works e.g. [Yang et. al. (2018)](https://arxiv.org/abs/1810.05866) and [Chen et. al. (2017)](https://arxiv.org/abs/1704.00763) who use both terms and were written around the same time period.

This disctinction is noted in [Rhodes (2018) - Not Enough Attention](https://iconix.github.io/dl/2018/07/06/not-enough-attention):

> While the original attention mechanism learns dependencies between a source and target sequence, self-attention learns dependencies within a single sequence. It computes the relevance and relationships of the other words in the sequence to the target word.

We can make this inference also by reading this [blog post](https://discourse.numenta.org/t/a-deeper-look-at-transformers-famous-quote-attention-is-all-you-need/10523/12):

> It’s called self-attention because the subspaces projected by the $W_q$, $W_k$, and $W_v$ transformations are performed on top of the same sequence, input sequence $[x_0, x_1, \dots ] \ \forall \ x \in  V_{50000}$. Because Query, Key and Value vectors are derived from the same initial tensor those (trainable) matrices compute a kernel for inter-relationships between ever $x_i$. The “self” is thus derived from computing that kernel onto itself.

The author then goes on to contrast self-attention with cross-attention:

> You “cross”-attend when the derivations of $Q$, $K$, and $V$ aren't similar. This is when you cross-attend the latents between the encoder and decoder to propogate information. 

In this light, I believe intra-attention is actually the most implicitly meaningful term based on what self-attention is described as doing, or to what the prefix "self-" is referring. 


**Note**: Using this definition, self-attention is not being defined by the architecture of the mechanism itself, but instead by the input parameters it is considering. As we see in Vaswani et al (2017) for example, the same attention mechanism is being used to provide intra and inter attention.

I have seen alternate claims that the use of "self" refers to the model using the attention mechanism and not a given token in the input sequence. As the attention mechanism is weighting the hidden states of the model relative to an input token, we can think of the model ass attending to itself. Through training, the model's attention mechanism learns how to attent to the model's hidden states, thus it learns how to a attend to the model itself.

I had seen claims that with self-attention, the only source of information is the input sequence. I understand the point that the author is trying to make, but I think the use of terminology is a bit careless; the hidden states of the model and the input sequence are inputs to the attention mechanism.

I have seen claims that dot-product attention is self-attention. After reading [Lin et. al. (2017)](https://arxiv.org/abs/1703.03130) I am sure that there are many types of attention and many types of self-attention; dot-product attention being onde of them.

### Inacurate And Misleading Statements

I have seen a number of blog posts that define self-attention according to the mathematical formula put forward by Viswani et. al. (2017). They claim that self-attention is also known as QKV-attention. I believe this is false; Viswani et. al. (2017) put forward a type of self attention not a restrictive definition for the implimentation of self-attention.

Despite what many claim, self-attention was not introduced or coined in Viswani et. al. (2017) as they acknowledge in that paper:

> Self-attention, sometimes called intra-attention is an attention mechanism relating different positions
of a single sequence in order to compute a representation of the sequence. Self-attention has been
used successfully in a variety of tasks including reading comprehension, abstractive summarization,
textual entailment and learning task-independent sentence representations [4, 22, 23, 19].

# History

## Ling et. al. (2015) - Self Attention Applied To Word Embeddings

In September 2015, Ling et. al. [published](https://aclanthology.org/D15-1161/) *Not all contexts are created equal: Better word representations with variable attention*.

In this paper, the authors describe an attention based extension to Continuous Bag Of Words (CBOW).

The attention mechanism mechanism calculates the importance of each word relative to every other word in the input sequence using a soft window applied to the entire vocabulary $V$. Becuase the scores are with respect to the other words in the input sequence, we can see that this is an intra-attention or self-attention mechanism.

Another helpful note on terminology: we can see that this attention mechanism is additive in that it sums the pairwise product of an attention vector and a CBOW word embedding vector.

According to [Lin et. al. (2017)](https://arxiv.org/abs/1703.03130) however, this approach has a shortcoming.

> Some recent work have also proposed supervised methods that use intra/self-sentence attention. Ling
et al. (2015) proposed an attention based model for word embedding, which calculates an attention
weight for each word at each possible position in the context window. However this method cannot
be extended to sentence level embeddings since one cannot exhaustively enumerate all possible
sentences.

**Note** The paper has an interesting connection with topic modeling where we see the authors discussing using attention to craft embeddings separately for seq2seq prediction and topic modeling:

> We introduce an extension to the bag-of-words model for learning words representations that take into account both syntactic and semantic properties within language. This is done by employing an attention model that finds within the contextual words, the words that are relevant
for each prediction. The general intuition
of our model is that some words are only
relevant for predicting local context (e.g.
function words), while other words are
more suited for determining global context, such as the topic of the document.

According to [Lin et. al. (2017)](https://arxiv.org/abs/1703.03130) also acknowledge this implimentation as self-attention:

> The LSTMN model (Cheng et al., 2016) also proposed a very successful intra-sentence level attention mechanism, which is later used by Parikh et al. (2016). We see our attention and theirs as having

recurrent iteration, which is sort of an ”online updating” attention. It’s more fine-grained, targeting
at discovering lexical correlations between a certain word and its previous words. On the contrary,
our attention mechanism is only performed once, focuses directly on the semantics that makes sense
for discriminating the targets. It is less focused on relations between words, but more on the semantics of the whole sentence that each word contributes to. Computationally, our method also scales up
with the sentence length better, since it doesn’t require the LSTM to compute an annotation vector
over all of its previous words each time when the LSTMN computes its next step.

## Liu et. al. (2016) - Inner Attention

In May of 2016, [published](https://arxiv.org/abs/1605.09090) *Learning natural language inference using bidirectional LSTM model and inner-attention*.

In the abstract, the authors define the purpose of their paper:

> In this paper, we proposed a sentence encoding-based model for recognizing text entailment.

Without going into too much detail, text entailment is an conceptual NLP model for the relationships between words which allows for inferences tobedrawn.

The authors go on to note that a key aspect to their approach is the use of a novel attention mechanism

> (An) attention mechanism was employed to replace average pooling on the same sentence for better representations. Instead of using target sentence to attend words in source sentence, we utilized the sentence’s first-stage representation to attend words appeared in itself, which is called ”Inner-Attention” in our paper .

According to [Linet. al. (2017)](https://arxiv.org/abs/1703.03130) however, this is different than self-attention

> Liu et al. (2016a) proposes a sentence level attention which has a similar motivation but
done differently. They utilize the mean pooling over LSTM states as the attention source, and use
that to re-weight the pooled vector representation of the sentence.


## Lin et. al. (2017) - Self-Attention Coined

In March of 2017, Lin et. al. [published](https://arxiv.org/abs/1703.03130) *A Structured Self-attentive Sentence Embedding*.

**Note**: I belive this is the origin for the term self-attention. This is the first paper I could find on the arxiv.org website that uses the term self-attention. All of the cited prior works do not use the term.

The authors propose a new model for extracting interpretable sentence embeddings. And this is the first major clue; the model is not deriving the embedding while performing downstream tasks like seq2seq etc. Although they note

> We evaluate our model on 3 different tasks: author profiling, senti-
ment classification and textual entailment.

A key differentiator is that they are looking at sentence embeddings rather than word or token embeddings. They note the shortcomings of the prior work with respect to embeddings.

> Much progress has been made in learning semantically meaningful distributed representations of
individual words, also known as word embeddings (Bengio et al., 2001; Mikolov et al., 2013).
On the other hand, much remains to be done to obtain satisfying representations of phrases and
sentences.

They classify existing approaches to sentence embeddings into two categories: generic sentence embeddings and task-specific sentence embeddings usually co-trained with the task performing model:

> Those methods generally fall into two categories. The first consists of universal sentence
embeddings usually trained by unsupervised learning (Hill et al., 2016). This includes SkipThought
vectors (Kiros et al., 2015), ParagraphVector (Le & Mikolov, 2014), recursive auto-encoders (Socher
et al., 2011; 2013), Sequential Denoising Autoencoders (SDAE), FastSent (Hill et al., 2016), etc.
>
> The other category consists of models trained specifically for a certain task. They are usually
combined with downstream applications and trained by supervised learning.

The authors then discuss the common approaches to deriving the embeddings. Ultimately a neural network (recursive, recurrent, convolutional, or LSTM) generates hidden states, either a sequential set of hidden states, or a final set of hidden states. The attention mechanism then weights these hidden states in it's computation of the attention score. As such, these hidden states are essentially word embeddings; they store latent representations of the input sequence.

> that processes the input sequence to produce a sequence of hidden state vectors. The question then becomes what to do next in order to get to the output classification problem. A common approach might be to use only the final hidden state (once the whole sequence has been processed) since this surely is encoding the context and content of the entire sequence. That final hidden state would then be a way of compressing an arbitrary-length sequence in a fixed-size vector and that vector could then be fed into a simple downstream layer for classification. Alternatively, one might aggregate the sequence of hidden state in some way (summation or averaging), though this would be lossy. 

Another notable distinction to their approach is the embedding itself:

> Instead of using a vector, we use a 2-D matrix
to represent the embedding, with each row of the matrix attending on a different
part of the sentence.

The authors continue to discuss key distinctions of their attention mechanism:

> Various supervised and unsupervised sentence embedding models have been mentioned in Section 1. Different from those models, our proposed method uses a new self-attention mechanism that
allows it to extract different aspects of the sentence into multiple vector-representations. The matrix
structure together with the penalization term gives our model a greater capacity to disentangle the
latent information from the input sentence. We also do not use linguistic structures to guide our
sentence representation model. Additionally, using our method we can easily create visualizations
that can help in the interpretation of the learned representations

according to this [post](http://keeganhin.es/blog/attn.html)

> they propose to use attention in order to embed abitrary-length sentences into a fixed vector representation. Further, they generalize the idea of attention and introduce the idea of multiple attention.
>
> They frame their model in the context of a typical NLP sequence classification task (sentiment analysis), but note that this task could be swapped out for any other. For sequence classification, their model begins in a common way: a bi-directional LSTM that processes the input sequence to produce a sequence of hidden state vectors. The question then becomes what to do next in order to get to the output classification problem. A common approach might be to use only the final hidden state (once the whole sequence has been processed) since this surely is encoding the context and content of the entire sequence. That final hidden state would then be a way of compressing an arbitrary-length sequence in a fixed-size vector and that vector could then be fed into a simple downstream layer for classification. Alternatively, one might aggregate the sequence of hidden state in some way (summation or averaging), though this would be lossy. These limitations feel very similar to problems we discussed with seq2seq, and Lin et al propose to use attention as a new way of handling this problem. However, note that unlike in seq2seq there is no output sequence of any kind - we only have a classification problem. So we are not attending to various words in order to align tokens between two sequences. Instead, we have only one input sequence and we might use attention over only that sequence in order to non-uniformly weight the components of the sequence so they can be combined into a fixed representation. For this reason, the term "self-attention" has become popular.


## Vaswani et al (2017) - Transformers and Attention
