# Overview

In this notebook we review the basic concepts realated to Retrieval Augmented Generation (RAG).

As discussed below, RAG allows pretrained models to get access to knowledge which is not contained in the model weights (i.e. the training set). This feature is valuable because it allows pre-trained models to be used in domain specific contexts or to address evolving information as it becomes available, without neeting to be retrained.

# Motivations

As the original authors note, while large pre-trained models store information in their parameters, their ability to access and manipulate knowledge is still limited. Additionally, they note that providing provenance for their decisions and updating their world knowledge remain open research problems.

As we will see, the RAG architecture helps address these issues.

# Major Topics

## Retrieval Augmented Generation (RAG)

[Lewis et. al. (2020)](#Lewis-et-al.-(2020)---RAG-Published) are credited as the creators of RAG. The novel paradigm allows LLMs to have access to knowledge which is not contained in the weights (i.e. which the model was not trained on). This greatly increases the power and flexibility of the models.

## RAG vs. Fine Tuning

Generally speaking, Fine Tuning and RAG are complementary; two tools solving related by separate problems. RAG allows an LLM to efficiently access relevant and/or updated information through in-context learning while Fine Tuning allows for task specific learning or specialization of the pre-trained LLM.

## Prompts and Prompt Engineering

The concept of prompt engineering is discussed in detail in the following [notebook](Prompt%20Engineering.ipynb).

Generally speaking, a prompt is the input secuence provided to an LLM and prompt engineering is the process of designing or manipulating the prompt such that the LLM returns the optimal output given a training set / input sequence.

The implimentation of a classical RAG or Agentic RAG system relies of the use and knowledge of prompt engineering.

## Types of Machine Learning Models And Training Techniques

Generally speaking, a machine learning model is used to predict an output given some input. Different types of ML models require different techniques to train or calibrate them (i.e. fit their parameters/hyperparameters to the data to get the best result). And broadly spreaking there are three major schools of training techniques for machine learning models. These being: supervised, unsupervised, and reinforcement learning. 

**Note**: As with this repository, it is common parlance to conflate the type of training a model requires with the model type. For example it would be equivalent to say "a model was trained using an unsupervised technique" and "the model is unsupervised" or "it's an unsupervised model".

**Note**: It is possible for a single model to be trained using multiple training techniques. For example a model might be trained with either supervised or reinforcement learning; or both; successively or in parallel (depending on the architecture of the model and training system). In certain scenarios, the combination of training techniques are given special names; for example "fine tuning" is the combination of supervised and reinforcement leanrning when training an LLM.

With supervised learning, the training set contains an input and a correct output; and the model then calibrates itself by learning a general pattern between input and output that roughly applies across the training set. With unsupervised learning, there is no output in the training set, because the model is expressing a mathematical relationship between the inputs and the model simply calibrates itself by learning the patterns between the different inputs. The "correctness" of the model's output is thus dependent only on the quality of the input data's representation of the distibution of the prediction space. 

### Reinforcement Learning

With Reinforcement learning, the training set also has no correct output included. Instead, the model learns how to make the correct prediction based off of a reward it is given during each training iteration by an interpreter (sometimes a human). A loss function is constructed that instructs the model how to adjust its parameters based on the reward issued by the interpreter.

As we will see, reinforcement learning is important tool for maximizing the performance of an LLM.

### Fine Tuning

Fine tuning refers to a process where a model is trained multiple stages. Where, in each stage of training, the same model different training techniques, and training sets, are used to adjust the parameters and hyperparameters of the model.  

The motivation for this technique is to facilitate practical specializations of the LLMs which are cost prohibitive to train from scratch. A common convention being that an LLM will first be trained using a supervised learning technique and large diverse corpus to get it's weights "close" to the optimal point and establish a baseline language model, and then the LLM will be trained in a second stage using a reinforcement learning technique and a task specific data set within the same domain (i.e. same language(s) or lexicon(s)) to give the model more specialized predictive power for the desired tasks.

With this segmentation of training phases, we are effectively performing [Transfer Learning](ExtraTopics.ipynb#Transfer-Learning-&-Pre-Trained-Models) between different use cases of the same domain.

There are many types of fine tuning techniques.

## Context Augmentaion

The term **Context Window** refers to the input sequence provided to an LLM. The context window is ultimately what conditions or determines the output of the LLM. Abstractly speaking, **Context Augmentation** refers to the process of inserting tokens into the context window such that the output of the LLM better reflects the user's desired output.

For example, if we were to ask the LLM a question, we might augment the question with a paragraph containing the answer in hopes that this conditions the LLM to return an output more aligned with the augmenting information. 

For more information see this [article](https://dev.to/emanuelferreira/understanding-context-augmentation-behind-llama-index-for-beginners-jdl).

## Graph RAG

A survey and formalization of GraphRAG techniques can be found in [Peng et al. (2004) ](https://arxiv.org/pdf/2408.08921).

The authors cite several motivations including:
- The ability to capture significant structured relational knowledge that cannot be represented through semantic similarity alone.
- The ability to avoid duplicate information bloating the context window leading to the "lost in the middle dilemma"
- The ability to grasp global information and perform Query-Focused Summarization (QFS)

They then go on to explain how GraphRAG retrieves triplets of information rather than chunks of preembedded information to inject into a context window to aid in the generation of a resposne.

## Lewis et al. (2020) - RAG Published

In May 2020, Lewis et al. [published](https://arxiv.org/abs/2005.11401) *Retrieval-Augmented Generation for Knowledge-Intensive NLP Tasks* to address several established limitations of LLMS. 

**Note**: This paper is connected to Meta and Meta is [often cited](https://research.ibm.com/blog/retrieval-augmented-generation-RAG) as being responsible for the creation of RAG.

The authors highlight the problem that generalized LLMs will likely not contain domain specific information, or up to date information, in their pre-trained model weights and thus their performance will always lag behind tasks-sepcific architectures.

> Large pre-trained language models have been shown to store factual knowledge
in their parameters, and achieve state-of-the-art results when fine-tuned on downstream NLP tasks. However, their ability to access and precisely manipulate knowledge is still limited, and hence on knowledge-intensive tasks, their performance
lags behind task-specific architectures. 

Additionally they note that currently, updating or expanding the information contained in the LLM's parameters is not always practical. In many cases the data set and wall time for the training is so expansive that this is a sheer impossibility.

**Note**: It took me a couple passes to read through the paper. Some of the phrases and word choices were a bit hard to digest. The authors use the term "pre-trained seq2seq transformer", "parametric memory", "generation models" or "generative models" or "large pre-trained language models" to refer to LLMs. Here the term "parametric" speaks to the parameters or weights of the underlying neural networks that makup the transformer.

To address this limitation, rather than retrain the generative model from scratch, the authors propose a fine-tuning methodology which will give a pre-trained LLM access to the domain specific information. 

The methodology requires the introducion of a "document retriever" into the solution design. This document retriever then logically subdivides into two major components which provide two important functions: a query encoder and a document index. The purpose of the query encoder will encode the query into an embedding and the purpose of the document index will store documents, which were previously converted into embeddings, and and index them via their symanitc similarity. This is outlined in figure 1 below.

> <center><img src="images\lewis_2020_rag_diagram.png"></center>
>
> Figure 1: Overview of our approach. We combine a pre-trained retriever (Query Encoder + Document Index) with a pre-trained seq2seq model (Generator) and fine-tune end-to-end. For query $x$, we use Maximum Inner Product Search (MIPS) to find the top-K documents $z_i$ . For final prediction $y$, we treat $z$ as a latent variable and marginalize over seq2seq predictions given different documents.


**Note**: The authors clarify that they used a [BART](BART.ipynb) variant as the generator.

The basic workflow of the solution at inference time is that when a user submits a query, the query will be converted into an embedding that is then used to identify the n-most similar document embeddings. Together the query, and the original query and the decoded relevant documents are provided to the generator (the LLM) as the input prompt.

During the fine-tuning session (i.e. the second training session), the retriever and generator are co-trained. The intention is that the generator learns how to generate desirable resposes to queryies based on the documents, and the document retriever learns how to recognize documents relevant to the query. 

**Note**: Several times during the article, the authors discuss components of the retriever as useing parametric and non-parametric memory within their proposed RAG architecture. The term "memory" is being used abstractly, and IMHO generously, to refer to language syntax information and domain specific information relevant to the generator's output. According to their paper, parametric memory refers to the trainable embedding model (which we can see is parameterized) and non-parametric memory refers to the document index.

The article goes on to outline the math behind the two models involved in their RAG implimentation. All of this is a bit out of scope for this notebook; but the important insight is that the models are all show to be conditioned based on the retrieved documents.

## Siriwardhana et al. (2022) - RAG-end2end

On October 6th, Siriwardhana et al. [published ](https://arxiv.org/abs/2210.02627) *Improving the Domain Adaptation of Retrieval Augmented Generation (RAG) Models for Open Domain Question Answering*. The abstract notes:

> Retrieval Augment Generation (RAG) is a
recent advancement in Open-Domain Question Answering (ODQA). RAG has only
been trained and explored with a Wikipediabased external knowledge base and is not
optimized for use in other specialized domains such as healthcare and news. In this
paper, we evaluate the impact of joint training of the retriever and generator components of RAG for the task of domain adaptation in ODQA. We propose RAG-end2end,
an extension to RAG, that can adapt to a
domain-specific knowledge base by updating all components of the external knowledge base during training. In addition, we
introduce an auxiliary training signal to inject more domain-specific knowledge. This
auxiliary signal forces RAG-end2end to reconstruct a given sentence by accessing the
relevant information from the external knowledge base. Our novel contribution is unlike
RAG, RAG-end2end does joint training of
the retriever and generator for the end QA
task and domain adaptation.

## Low (2023) - RAG without Vector DB

On December 23rd 2023, Yucheng Low [published](https://about.xethub.com/blog/you-dont-need-a-vector-database) an article titled *You Don't Need a Vector Database*. As the name suggests, the article makes the case that RAG can be performed without having to stand up a vector database. The basic idea is that RAG will be performed by two filters: first a keyword search returns the N most relevant document chunks; then symantic search is performed on the coresponding vectors locally. With this approach, the more efficient symatic search is performed on the global knowledge set while the symantic search is only performed on the filtered subset of information. With this approach, Low asserts that the precomputed embedding vectors can be stored in a traditional key-value store rather than a specialized vector database, as the computations for symantic similarity are small enough to be performed locally by the client rather than by the vector database.

**Note**: At this point in time, just-in-time embedding generation is not practical as the generative algorithms are not fast enough. For more information on this topic see the [Word Embeddings Notebook](Word%20Embeddings.ipynb#Benchmarks-For-Embeddings).

To substantiate his claims, Low provides the following diagram to highlight the performance gains of combining the tranditional keyword search using the BM25 algorithm and the newer sematic search using embedding vectors provided by OpenAI.

On March 14th, 2024, Cobus Greyling published an article on several sites ([e.g.](https://cobusgreyling.substack.com/p/agentic-rag-context-augmented-openai), [e.g.](https://cobusgreyling.medium.com/agentic-rag-context-augmented-openai-agents-578e96212bc0])) claiming that "LlamaIndex has coined the phrase Agentic RAG".



According to is [homepage](https://docs.llamaindex.ai/en/stable/), LlamaIndex is a framework for building context-augmented LLM applications. In other words, it provides the developer a way to build applications which rely on context augmentation.

> Some popular use cases include the following:
>
> - Question-Answering Chatbots (commonly referred to as RAG systems, which stands for "Retrieval-Augmented Generation")
> - Document Understanding and Extraction
> - Autonomous Agents that can perform research and take actions
>
> LlamaIndex provides the tools to build any of these above use cases from prototype to production. The tools allow you to both ingest/process this data and implement complex query workflows combining data access with LLM prompting.
