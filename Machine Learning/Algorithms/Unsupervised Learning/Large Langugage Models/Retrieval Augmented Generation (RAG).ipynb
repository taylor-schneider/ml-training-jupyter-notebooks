## Types of Machine Learning Models And Training Techniques

Generally speaking, a machine learning model is used to predict an output given some input. Different types of ML models require different techniques to train or calibrate them (i.e. fit their parameters/hyperparameters to the data to get the best result). And broadly spreaking there are three major schools of training techniques for machine learning models. These being: supervised, unsupervised, and reinforcement learning. 

**Note**: As with this repository, it is common parlance to conflate the type of training a model requires with the model type. For example it would be equivalent to say "a model was trained using an unsupervised technique" and "the model is unsupervised" or "it's an unsupervised model".

**Note**: It is possible for a single model to be trained using multiple training techniques. For example a model might be trained with either supervised or reinforcement learning; or both; successively or in parallel (depending on the architecture of the model and training system). In certain scenarios, the combination of training techniques are given special names; for example "fine tuning" is the combination of supervised and reinforcement leanrning when training an LLM.

With supervised learning, the training set contains an input and a correct output; and the model then calibrates itself by learning a general pattern between input and output that roughly applies across the training set. With unsupervised learning, there is no output in the training set, because the model is expressing a mathematical relationship between the inputs and the model simply calibrates itself by learning the patterns between the different inputs. The "correctness" of the model's output is thus dependent only on the quality of the input data's representation of the distibution of the prediction space. 

### Reinforcement Learning

With Reinforcement learning, the training set also has no correct output included. Instead, the model learns how to make the correct prediction based off of a reward it is given during each training iteration by an interpreter (sometimes a human). A loss function is constructed that instructs the model how to adjust its parameters based on the reward issued by the interpreter.

As we will see, reinforcement learning is important tool for maximizing the performance of an LLM.

## Fine Tuning

Fine tuning refers to a process where a model is trained multiple stages. Where, in each stage of training, the same model different training techniques, and training sets, are used to adjust the parameters and hyperparameters of the model.  

The motivation for this technique is to facilitate practical specializations of the LLMs which are cost prohibitive to train from scratch. A common convention being that an LLM will first be trained using a supervised learning technique and large diverse corpus to get it's weights "close" to the optimal point and establish a baseline language model, and then the LLM will be trained in a second stage using a reinforcement learning technique and a task specific data set within the same domain (i.e. same language(s) or lexicon(s)) to give the model more specialized predictive power for the desired tasks.

With this segmentation of training phases, we are effectively performing [Transfer Learning](ExtraTopics.ipynb#Transfer-Learning-&-Pre-Trained-Models) between different use cases of the same domain.

There are many types of fine tuning techniques.

As discussed where??, recall that an LLM is a model trained to predict (or generate) the next token of a sequence given the previous tokens in that sequence. By putting this functionality inside a loop, the model is able to generate a new sequence altogether given an input sequence.

Because of their size (i.e. the number of parameters or weights in its neural networks) an LLM can learn a huge set of statistical relationships between terms found in the corpus. And because of the attention mechanism built into the Transformers, which conditions these relationships based on context, the LLM is able to accurately model multiple different relationships between the same set of words and selectively generate using a statistical relationship that is most relevant to the input sequence.

This feature can be exploited and lies at the heart of an LLM's flexability and versatility. If trained on a sufficiently large and diverse corpus of text (text containing summaries, translations, categorizations, songs, etc.), the LLM's size will allow the model to learn to perform a wide range of NLP tasks rather than needing a single model for each specific task was was the case in the previous era of the small language models.

### GPT - GPT-3

As noted in the article [The Making of ChatGPT: From Data to Dialogue - Johri (2023)](https://sitn.hms.harvard.edu/flash/2023/the-making-of-chatgpt-from-data-to-dialogue) we can see this phenomena demonstrated by the evolution of the GPT-4 model:

> in 2018 ... GPT ... was capable of completing sentences and paragraphs. Over the next two years, they improved this model by training it on ... a massive dataset that contained many examples of specific tasks (such as answering questions, translating languages, etc.) ...and further increasing the model size. They also made engineering changes to the model architecture that enabled it to learn faster and more effectively than before. They finally achieved a 175 billion parameter model that they called GPT-3. ... GPT-3 was not trained for a specific task (such as translating languages or summarizing text), it was only trained to predict the next word. But since it was trained on a massive dataset ... it acquired the ability to perform these wide varieties of tasks when prompted appropriately.

However, when only being trained using the tranditional supervised learning method, GPT-3 was not able to understand any and all NLP instructions that were passed through the input sequence. For example when asked "Write a short story about the starts and moon" the model instead output random phrases and words. 

To further enhance the model's performance, the researchers added a second phase of training which used a reinforcement learning technique that relies of human feedback, known as RLHF for short. This second phase of training, referred to as fine tuning, allows the model to specialize for a particular task. In this case, the researchers guided the model's reinforced training session to allow the model to acheive higher performance scores. This model was called **InstructGPT**, and was released in 2022. 

### OpenAI (2022) - ChatGPT

According to [Johri (2023)](https://sitn.hms.harvard.edu/flash/2023/the-making-of-chatgpt-from-data-to-dialogue), with the next major evolution of the GPT series, the OpenAI team wanted to teach the model to be conversational. In other words, they wanted to teach the model to understand how to participate in conversations. To do this, they needed to orchestrate a training scenario where the input sequence would be the ongoing conversation and the output serquence would be the model's generated reaction to the input sequence. By manipulating the input sequence the engeers could dictate the model's position in the conversation. In a chat bot scenario this would allow them to dictate whether the model was the "human" who initiates the first response or the "bot" who responds to the input sequence. 

Fine tuning was again used for this effort. According to the article *[The inside story of how ChatGPT was built from the people who made it -  Heaven (2023)](https://www.technologyreview.com/2023/03/03/1069311/inside-story-oral-history-how-chatgpt-built-openai/)* the same base model from InstructGPT was fine tuned using a similar methodology as before (i.e. RLHF). However they added conversational data and tweaked the training process slightly. 

> 
> John Schulman: The raw technical capabilities, as assessed by standard benchmarks, don’t actually differ substantially between the models, but ChatGPT is more accessible and usable.

This work pits multiple chatbots against each other: one chatbot plays the adversary and attacks another chatbot by generating text to force it to buck its usual constraints and produce unwanted responses. Successful attacks are added to ChatGPT’s training data in the hope that it learns to ignore them.

Thinking about the use case of a chat bot for example, the user will enter text into the UI and the text will be inserted into a prompt which instructs the LLM that the text contains a question and the LLM should provide an answer. Similarely, for a summarization bot, the user would again enter text, which gets wrapped up inside a large text serquence which contains instructions telling the LLM that is should provide a summary of the text.
