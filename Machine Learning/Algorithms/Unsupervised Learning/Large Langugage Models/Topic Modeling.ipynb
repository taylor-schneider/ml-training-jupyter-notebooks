# Overview

Topic modeling, as the name suggests, is concerned with using a machine learning model to understand what topics are being discussed or addressed in a corpus of text. These could be documents, conversations between people, or even chat bots. Additionally, it's possible that a body of text would imply an interest in multiple topics, this topic modeling would be concerned about identifying which topics and their relative relevance to a given segment of text and the proportionality relative to other identified topics for the same segment. 

# History

## Maron (1961) - Beginning Of Auto Classification

In 1961, Maron [published](https://dl.acm.org/doi/pdf/10.1145/321075.321084) *Automatic Indexing An Experimental Inquiry* which as far as I can tell, is the beginning of the field of automatic classification. The paper proposes a novel and rigorous approach to classifying documents:

>  The task, in essence, is to have a computing
machine read a document and on the basis of the occurrence of selected clue words decide to
which of many subject categories the document in question belongs. This paper describes
the design, execution and evaluation of a modest experimental study aimed at testing
empirically one statistical technique for automatic indexing. 
>
> ...
>
> This approach to the problem of automatic indexing is a
statistical one. It is based on the rather straightforward notion that the individual words in a document function as clues, on the basis of which a prediction
can be made about the subiect category to which the document most probably
belongs. The fundamental thesis says, in effect, that statistics on kind, frequency,
location, order, etc., of selected words are adequate to make reasonably good
predictions about the subject matter of documents containing those words. We
do not consider the meanings of individual words--we only look to see what
subjects they describe and from such data make predictions. 
>
> ...
>
>  Given this approach to automatic indexing, two
problems present themselves, viz., the selection of clue words and the prediction
techniques relating clue words and subject categories.


The paper goes on to describe the clue selection process and the rigorous mathematical framework for the probablistic model used to do the classification. In a nutshell, the selection of clue words is done manually, but is based on descriptive statistical proprties attributed to the clue word by analyzing the corpus as a whole. Things like rarity, prevelance through multiple documents, and other considerations were used to assess the information contained in a prospective clue word and it's usefulness to the prediction framework. The prediction framework relies on a conditional distribution with an a priori assumption that Shannon's expression for entropy will model the uncertainty attached to a clue word. Using this assumption and some other assertions derived from the relationship of a term, a document, and a corpus, a set of conditional expressions are proposed to model the probability that a document will be modeled under a given category.  

This paper marks a new generatation of classification techniques, not only for it's use of computers but also for it's use of a rigorous probabalistic model to perform the classification. We will see that the papers to follow this publication differ in the implimentation, but the overall theory remains largely unchanged.

## Borko and Bernick (1963, 1964) - Factor Analysis For Document Classification

In 1963, Borko and Bernick [published](https://dl.acm.org/doi/pdf/10.1145/321160.321165) *Automatic Document Classification* followed by a second publication, *Autolnatic Document Classification Part II. Additional Experiments* which was [published](https://dl.acm.org/doi/pdf/10.1145/321217.321219) in 1964.

In this series, the authors discuss using a factor analysis technique to automatically derive classifications and then classify documents based on the text contained in their abstract.

**Note**: Factor analysis is a mathematical technique which is used to reduce a large number of observed variables (i.e. dimensions) into a smaller number of unobserved (theoretical) variables called factors. The technique assumes that there is a covariance between observed variables which can be represented in a smaller dimensional space than the observed space. The methematical technique works by modeling the observed variables using linear regression models where the terms of the model are the unobserved factors. A wikipedia article on the technique can be found [here](https://en.wikipedia.org/wiki/Factor_analysis).

## Deerwester et. al. (1990) - Latent Symantic Indexing Published

In 1990, Deerwester et.al. [published](http://wordvec.colorado.edu/papers/Deerwester_1990.pdf) "Indexing by latent semantic analysis" which proposes a novel approach to indexing and retrieval of text based resources. The paper's abstract contains the following statements:

> The approach is to take
advantage of implicit higher-order structure in the association of terms with documents ("semantic
structure") in order to improve the detection of relevant documents on the basis of terms found in
queries. The particular technique used is singular-value decomposition, in which a large term by
document matrix is decomposed into a set of ca 100 orthogonal factors from which the original
matrix can be approximated by linear combination. Documents are represented by ca 100 item
vectors of factor weights. Queries are represented as pseudo-document vectors formed from
weighted combinations of terms, and documents with supra-threshold cosine values are returned.

**Note**: I believe the term *ca* is a mathematical shorthand for the latin word *circa* meaning around or approximately. For more information of equivalent abreviations see this [paper](https://www.math.ucdavis.edu/~anne/WQ2007/mat67-Common_Math_Symbols.pdf).

The motivation for the proposal is due to the shortcomings of the indexing and retrieval methods of the time. As stated in this [paper](https://www.cse.msu.edu/~cse960/Papers/LSI/LSI.pdf):

> Typically, information is retrieved by literally matching terms in documents with
those of a query. However, lexical matching methods can be inaccurate when they
are used to match a user's query. Since there are usually many ways to express a
given concept (synonymy), the literal terms in a user's query may not match those
of a relevant document. In addition, most words have multiple meanings
(polysemy), so terms in a user's query will literally match terms in irrelevant
documents.
> ...
> Latent Semantic Indexing (LSI) [Deerwester et al] tries to overcome the
problems of lexical matching by using statistically derived conceptual indices
instead of individual words for retrieval. LSI assumes that there is some
underlying or latent structure in word usage that is partially obscured by
variability in word choice.

The underlying or latent structure in word usage is exposed through spectral analysis.

**Note**: A spectrum is a continuous sequence or range. A variable is typically defined as operating on or in a domain or spectrum. The plural of spectrum is spectra. Thus a multidimensional variable models one or more spectrums i.e. a spectra. Spectral analysis is the study of a spectrum or spectra. In the context of data science, Spectral Analysis is often implicitly connected to the measurement of eigenvalues and eigenvectors. Both Single Value Decomposition and Principal Component Analysis (PCA) are techniques which transform a observed high-dimensional space into a theoretical lower-dimensional space via eigenvectors and quantify the information captured in the eigenvectors via the coresponding eigenvalues. As such both SVD and PCA are considered tools for spectral analysis.

Deerwester et.al. use SVD to estimate the latent structure of word usage accross documents. Indexing and retrieval are then performed using the eigenvalues and eigenvectors obtained from the SVD.

With this approach, documents are broken down into a set of terms which are represented as vectors. This representation is referred to in the paper as the term-document matrix. The vectors (i.e. the matrix) are then projected into a new low-dimensional space using the single value decomposition (SVD) technique. The Eigenvalues produced through the decomposition then help identify the most striking correlations between terms in the matrix. 
 

> Latent semantic indexing (LX) is an information retrieval technique based on the spectral analysis of the
term-document matrix, whose empirical success had heretofort been without rigorous prediction and explanation.
We prove that, under certain conditions, LSI does succeed in capturing the underlying semantics of the corpus
and achieves improved retrieval performance. We also
propose the technique of random projection as a way of
speeding up LSI. We complement our theorems with encouraging experimental results. We also argue that our
results may be viewed in a more general framework, as
a theoretical basis for the use of spectral methods in a
wider class of applications such as collaborative filtering. 
>
> Papadimitriou et. al. (1998)


**Note**: While the technique of using a multidimensional vector to represent semantic meaning is similiar conceptually to that of word embeddings, these two techniques are very different under the hood and arose from different fields of study. For more information see the complementary [Word Embeddings Notebook](Word%20Embeddings.ipynb).

**Note**: The use of LSI and Latent Semantic Analysis (LSA) seem to be interchangable in the literature.

## Papadimitriou et. al. (1998) - Adding rigor to LSI

In May of 1998, Papadimitriou et. al. [published](https://dl.acm.org/doi/pdf/10.1145/275487.275505) "Latent semantic indexing: a probabilistic analysis" which proposes a rigorous explanation for why LSI works and provides a methematical framework which explains or predicts performance improvements to LSI implimentations.

the convex combination of a small number of topics.

the term distribution. A corpus is then a collection of

probabiliitic model, motivated in part by applications
to collaborative filtering.) 
>
> Once we have a corpus model, we would like to determine under what conditions LSI results in enhanced
retrieval, We would like to prove a theorem stating
essentially that if the corpus is a reasonablyfocused collection of meaningfully correlated documents, then LSI
performs urell. The problem is to define these terms
so that (1) there is a reasonably close correspondence
with what they mean intuitively and in practice, and
(2) the theorem can be proved. In Section 4 we prove
results that, although not quite as general as we would
have liked, definitely point to this direction. In particular, we show that in the special case in which (a) there
is no style modifier; (b) each document is on a single

on its own terms, and low probability on all others; then

number of topics, wilI discover these topics exactly, with
high probability (Theorem 2). 
>
> In Section 5 we show that, if we project the termdocument matrix on a completely random low-dimensional
subspace, then with high probability we have a distancepreservation property akin to that enjoyed by LSI. This
suggests that random projection may yield an interesting improvement on LSI: we can perform the LSI precomputation not on the original term-document matrix,
but on a low-dimensional projection, at great computational savings and no great loss of accuracy (Theorem 4). 
>
> Random projection can be seen as an alternative to
(and a justification of) somplingin LSI. Reports on LSI
experiments in the literature seem to suggest that LSI
is often done not on the entire corpus, but on a randomly selected subcorpus (both terms and documents
may be sampled, although it appears that most often
documents are). There is verv little non-emnirical evidence of the acuracy of such Isampling, Our iesult suggests a different and more elaborate (and computationally intensive) approach -projection on a random lowdimensional subspace- which can be rigorously proved

## Hofmann (1999). Probabilistic Latent Semantic Indexing (PLSI/PLSA)

In 1999, while researching at berkley, Hofmann [published](https://web.archive.org/web/20101214074049/http://www.cs.brown.edu/~th/papers/Hofmann-SIGIR99.pdf) a paper called *Probabilistic Latent Semantic Indexing*. Hofman proposes a novel improvement to the prior work which uses word count to derive latent syntactic classifications. His approach is to optimize a generalization of the Expectation Maximization algorithm using a set of training data. The abstract notes:

> the
utilized model is able to deal with domain{specific synonymy
as well as with polysemous words. In contrast to standard
Latent Semantic Indexing (LSI) by Singular Value Decomposition, the probabilistic variant has a solid statistical foundation and defines a proper generative data model. Retrieval
experiments on a number of test collections indicate substantial performance gains over direct term matching methods as well as over LSI.

Hoffman goes on to state:

> Although LSA has been applied with remarkable success
in dierent domains including automatic indexing (Latent
Semantic Indexing, LSI) [1, 3], it has a number of deficits,
mainly due to its unsatisfactory statistical foundation. The
primary goal of this paper is to present a novel approach
to LSA and factor analysis { called Probabilistic Latent Semantic Analysis (PLSA) { that has a solid statistical foundation, since it is based on the likelihood principle and de-
fines a proper generative model of the data. This implies in
particular that standard techniques from statistics can be
applied for questions like model fitting, model combination,
and complexity control. In addition, the factor representation obtained by PLSA allows to deal with polysemous words and to explicitly distinguish between different meanings and different types of word usage.

Additionally he notes that in prior works, the statistical model for the general cooccurance data which associated observed and unobserved variables is referred to as an aspect model:

> The core of PLSA is a statistical model which has been called
aspect model [7, 15].

And gives the following sources:

## Pritchard et. al. (2000) - Latent Dirichlet Allocation (LDA)

In the context of population genetics, LDA was proposed by J. K. Pritchard, M. Stephens and P. Donnelly in 2000 via a [paper](https://academic.oup.com/genetics/article/155/2/945/6048111?login=false) titled *Inference of population structure using multilocus genotype data*.

## Blei et. al. (2003) - LDA applied to ML

In Januaray 2003, Blei et. al. [published](https://web.archive.org/web/20120501152722/http://jmlr.csail.mit.edu/papers/v3/blei03a.html) *Latent Dirichlet Allocation* in the Journal of Machine Learning Research. This is the earliest application of LDA within the machine learning field that I could find.

According to the abstract: 

> We describe latent Dirichlet allocation (LDA), a generative probabilistic model for collections of discrete data such as text corpora. LDA is a three-level hierarchical Bayesian model, in which each item of a collection is modeled as a finite mixture over an underlying set of topics. Each topic is, in turn, modeled as an infinite mixture over an underlying set of topic probabilities. In the context of text modeling, the topic probabilities provide an explicit representation of a document. We present efficient approximate inference techniques based on variational methods and an EM algorithm for empirical Bayes parameter estimation. We report results in document modeling, text classification, and collaborative filtering, comparing to a mixture of unigrams model and the probabilistic LSI model.

**Note**: In this context the term *EM algorithm* refers to the variational Bayesian Expectation–Maximization algorithm. This is closely related to the Maximum Liklihood Estimation (MLE) technique. MLE gives you the objective to maximize. EM is one method for finding this maximum. In other words, EM finds (or attempts to find) the MLE iteratively. There are other methods besides EM that can be applied to find the MLE.

Several extentions to the original LDA have been proposed to advance the field of topic modeling. For example PAM and HPAM as will be discussed below.

## Li & McCallum (2006) - Pachinko Allocation Model (PAM)
In 2006, Li et. al. [published](https://people.cs.umass.edu/~mccallum/papers/pam-icml06.pdf) a paper titled *Pachinko allocation: DAG-structured mixture models of topic correlations* which proposes a novel extension to the LDA model in order to additionally model the relationships between topics, which was previously being ignored. The abstract states:

> LDA does
not capture correlations between topics. In
this paper, we introduce the pachinko allocation model (PAM), which captures arbitrary,
nested, and possibly sparse correlations between topics using a directed acyclic graph
(DAG). The leaves of the DAG represent individual words in the vocabulary, while each
interior node represents a correlation among
its children, which may be words or other interior nodes (topics). PAM provides a flexible alternative to recent work by Blei and
Lafferty (2006), which captures correlations
only between pairs of topics.

## Li et. al. (2007) - Hierarchical PAM (HPAM)

In 2007, the PAM was extended by Li, McCallum, and David Mimno. Their [paper](https://mimno.infosci.cornell.edu/papers/icml-hpam.pdf) titled *Mixtures of hierarchical topics with Pachinko allocation* proposes an enhancement to PAM that explicitly represents a topic hierarchy. The abstract states:

> The four-level pachinko allocation model
(PAM) (Li & McCallum, 2006) represents
correlations among topics using a DAG structure. It does not, however, represent a
nested hierarchy of topics, with some topical word distributions representing the vocabulary that is shared among several more
specific topics. This paper presents hierarchical PAM—an enhancement that explicitly represents a topic hierarchy. This model
can be seen as combining the advantages of
hLDA’s topical hierarchy representation with
PAM’s ability to mix multiple leaves of the
topic hierarchy. Experimental results show
improvements in likelihood of held-out documents, as well as mutual information between
automatically-discovered topics and humangenerated categories such as journals.

## Liu et. al. (2017)

Other topic models are generally extensions on LDA, such as Pachinko allocation, which improves on LDA by modeling correlations between topics in addition to the word correlations which constitute topics. 

Hierarchical latent tree analysis (HLTA) is an alternative to LDA, which models word co-occurrence using a tree of latent variables and the states of the latent variables, which correspond to soft clusters of documents, are interpreted as topics.
