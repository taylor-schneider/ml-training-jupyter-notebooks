These could be documents, conversations between people, or even chat bots. Additionally, it's possible that a body of text would imply an interest in multiple topics, this topic modeling would be concerned about identifying which topics and their relative relevance to a given segment of text and the proportionality relative to other identified topics for the same segment.

Thus a multidimensional variable models one or more spectrums i.e. a spectra. **Spectral Analysis** is the study of a spectrum or spectra. 

### PCA, EFA, and Matrix Decomposition
In the context of data science, Spectral Analysis is often implicitly connected to the measurement of eigenvalues and eigenvectors. As such, the field is implicitly related to analytical techniques which deal with eigenvalues/eigenvectors (e.g. PCA or EFA). Confusingly, these analytical techniques are also confused/conflated with their accompanying mathematical techniques for performing matrix decomposition and deriving the eigenvalues/eigenvectore (e.g. SVD or LU Decomposition). 

Exploratory Factor Analysis (EFA) is a mathematical technique which can be used to reduce a large number of observed variables (i.e. dimensions) into a smaller number of unobserved (theoretical) variables called factors. The technique assumes a causal relationship between a set of unobserved variables and the observed variables. It continues to suppose that the through mathematical manipulation, the covariance can be decomposed so that the variance of the unobserved variables is estimated. The methematical technique works by modeling the observed variables using linear regression models where the terms of the model are the unobserved factors. A wikipedia article on the technique can be found [here](https://en.wikipedia.org/wiki/Factor_analysis).

Mathematically, Factor Analysis is similar to [PCA](../../../../Data%20Science/Principal%20Component%20Analysis%20(PCA)/Principal%20Component%20Analysis.ipynb) in that a feature correlation matrix is decomposed into eigenvalues which express the amount of information (variance) expressed by each feature. 

However the philosophy behind the use cases for the two methods has striking differences. With PCA, the causal relationship between the features and the principal components is that the features cause the variance in the principal components. However with Factor Analysis, the factors are viewed as causing the variance in the features. In other words, the causal relationships modeled by these techniques are understood and expected to be oposite from each other. [Tipping (1997)](https://publications.aston.ac.uk/id/eprint/38367/1/NCRG_97_010.pdf) provides the mathematical explanation of this difference.

A great article on the comparison of the two can be found [here](https://community.jmp.com/t5/JMP-Blog/Principal-components-or-factor-analysis/ba-p/38347).

A similar stackexchange article can be found [here](https://stats.stackexchange.com/questions/200250/can-svd-be-used-to-perform-factor-analyis).

I think these two articles with Tipping (1997) should provide a clear and complete explanation

## MLE and EM

In many applied sciences, data scientists use probabilistic models to make predictions for an observed system. They attemt to build and tune a model to make predictions which are consistent with the observed values. As the name suggests, these probabilistic models rely on probability distributions. Distrubutions inherently are defined by their distribution function which in turn accepts parameters. By varrying the values assigned to parameters, the probabilities ascribed to events by the distribution function change. In order for the distribution to accurately model the observed values, the distribution (and thus the model) must be calibrated. Calibration refers to the process of selecting parameters which minimize the difference between the model's distribution and the distribution of the observed events.

The maximum likelihood estimation (MLE) is a calibration method; i.e. a method of estimating the parameters of an assumed probability distribution. As the name suggests, the method is built around the concept of likelihood. While probability describes an expectation relative to a spceific outcome, likelihood describes the expectation relative to a population or set of possible outcomes. Thus likelihood describes the expectation of a population parameter given an observation. Liklihood is related to probabilitiy, but often confused with it.

The goal of maximum likelihood estimation is to determine the set of distribution parameters for which corespond to the highest joint probability (i.e. likelihood) for the observed data.

There are many techniques or algorithms for performing the calibration. They all begin first by defining the likelihood function (i.e. the function which tells us the likelihood of a distribution having a particular parameter, which caused the observed events. 

In some cases the likelihood function is solvable analytically. For example in the case of a linear regression models with random error terms, the ordinary least squares (OLS) method can yield a solution.

In some cases the likelihood function is not solvable analytically. In such cases, numerical methods can be used to approximate the maxima of the likelihood function. These include gradient discent, newtons method, Expectation-maximization (EM), etc.

## Aspect Model

I believe the term was coined by Hoffman et. al. in the 1999 [publication](https://proceedings.neurips.cc/paper_files/paper/1998/file/0c8ce55163055c4da50a81e0a273468c-Paper.pdf) titled *Learning from Dyadic Data*.

Hoffman et. al. provide a rigorus and abstract mathematical definition for the term. An aspect model is defined in the context of dyadic data; data being composed of, or observations being made about dyads; elements selected from a domain which consists two finite sets. Thus a dyad is a pair with one element from either set.

He notes some tangible examples of use cases where the domain would consist of dyadic data including the example of Text-based information retrieval:

> where $X$, may correspond to a document collection, $Y$ to keywords, and $(x_i, y_k)$ would represent the occurrence of a term $y_k$ in a document $x_i$.

In the paper, Hoffman et. al. present a general framework for performing unsupervised learning on dyadic data. After defining dyadic data, he begins discusing the two common tasks central to unsupervised learning relating to dyadic data: 

> (i) probabilistic modeling, i.e., learning a joint or conditional probability model over $X \times Y$, and (ii) structure discovery, e.g., identifying clusters and data hierarchies.


Hoffman then goes on to enumerate the four ways in which latent class models can be defined. The first of which is the aspect model

1. **Aspect based**

    Introduce an onobserved mapping $c$ which partitions the dyadic domain $X \times Y$ into $K$ classes.

    $$ X = \{ x_1 ... x_N\}, \ Y = \{ y_1 ... y_M\}$$
   $$ c : X \times Y \rightarrow \{c_l , . . . , c_K \}$$

    The pre-image (i.e. inverse transform) of the observed dyad $c_k$ is referred to as an aspect $a_k$.

   $$ c^{-1}(c_k) = a_k = (x_n, y_m)$$

    The set of observed dyads being the sample set $S$.

    
    $$ (x_n, y_m) \in S $$

So we see that the aspect is the element from the domain which, through the unobserved mapping, caused or generated the observed dyad. Additionally we see that in this context, an aspect model, as the name suggests, is a model for how aspects map to dyads.

Hoffman then goes on to add additional rigor to the definition of an aspec model and notes the following assumptions:

* all co-occurrances $ (x_n, y_m) \in S $ are i.i.d
* $x_n$ and $y_m$ are conditionally independent given the class $k$

Hoffman then derives a formulas for the posterior probabilities using the Expectation Maximization (EM) approach for Maximum Likelihood Estimation (MLE) and credits Dempseter et. al. 1977 for the derevation.

## Maron (1961) - Manually Classifying Documents via Keyword Frequency

In 1961, Maron [published](https://dl.acm.org/doi/pdf/10.1145/321075.321084) *Automatic Indexing An Experimental Inquiry* which as far as I can tell, is the beginning of the field of automatic classification. 

The paper proposes a novel and rigorous approach for automatically classifying documents. 

After reviewing the paper I belive the high level process is still being used to some degree by LLM models and other NLP techniques. Below is a high level summary of the sequential steps Maron followed:
1. Assemble a corpus of documents
1. Select a set of categories
1. Map documents to categories
1. Partition corpus into two sets
1. Analyze word occurance correlation with categorization for group 1
1. Selet keywords to use as predictors
1. Define and calibrate the predictive Model
1. Classify document in group 2 using keywords
1. Compare predicted classification with Analyze results

**Assemble a corpus of documents**

In 1951, the Institute of Radio Engineers (IRE) formed its Professional Group on Electronic Computers (PGEC) [[source](https://www.computer.org/about/cs-history)]. In March 1959, the PGEC inagurated a new abstracting service which provided abstracts and summaries for a wide variety of publications in the computer science field.

Maron used the 405 abstracts provided by PGEC as the corpus of documents on which to conduct his experiment. 

**Select a set of categories**

After choosing the corpus, Maron needed to define the categories that documents would be sorted into. As noted in the acknowledgements, Maron selected 32 categories based on the documents in group 1 with the help of Michael Warshaw and then manually indexed (i.e. mapped or categorized) the documents to those categories.

**Map documents to categories**

Maron did not provide a detailed exlanation of how the categories were chosen for this experiment but did specify the 32 that were chosen. I am inferring that these were chosen abritratily.

> The PGEC had provided with its abstracts a classification 
system of 10 major categories, five of which were subdivided into about three 
subcategories each. However, it was decided to use a different and finer set of 
32 categories.

**Partition corpus into two sets**

Maron then performed the typical "train and test split" on the corpus that we see in modern machine learning techniques

> The complete corpus consisted of 405 different documents 
and was divided into two groups which we called group 1 and group 2. Group 1 
contained those 260 abstracts that appeared in the March and June issues of the 
1959 PGEC and it was the basis for the statistical data necessary to make the 
subsequent predictions. Group 2 consisted of those 145 abstracts which appeared 
in the September 1959 issue of the PGEC and was not even looked at during the 
preparatory phase of the experiments. Once the data from group 1 had been 
collected, the notions were tested by having a machine index the documents of 
group 2. 

Group 1 is used to essentially calibrate or train the classification model while Group2 is used to analyze the performance of the trained model.

**Analyze word occurance correlation with categorization**

In order to derive the statictical predictors and calibrate the classification model, the upfront word analysis needed to be conducted for each document. The frequency of word occurance was computed for each document so that subsequent calculations could be performed.

**Select keywords to use as predictors**

Maron does not use the term keyword, instead he elects to use "clue word". As the name suggests, he notes that the "clue words" implicitly provide suggestions as to the correct category. 

In section 5, Maron descibes a set of rules or arguments for how the set of all words in the corpus was filtered to the subset of "clue words" which provide statistically significant relationships with the categories. In a nutshell, Maron removes words that occur too frequently and elects words who's frequeny peaked with respect to a given category. Ultimately 90 clue words were selected from the set of ~3200 unique words.

Maron uses a matrix, similar to what is today referred to as a Document-Term Matrix, to keep track of the occurances of words relative to categories. He called this the "clue word-category" matrix. He likely also uses a document-term matrix as he discusses his analysis of how many clue words were contained in each document and how many occurrences of an ordinary word occurred in a particular document.

**Define and calibrate the predictive Model**

This is where Maron introduces a rigorous mathematical framework into the discussion. He uses a bayesian approach which establishes a conditional relationship desribing the probability that a word will be assigned a specific classification given the presence of a word. The probabilities of the model are estimated huristically by computing the fractions of observed quantities. The probability of being classified as $C_i$ for example is computed as the total number of documents categorized as $C_i$ divided by the total number of documents.

A key assumption of the model is that the categories selected are exclusive, exhaustive, and independent. Maron acknowledges however that this is not the case in reality:

> They are exhaustive in that every document must belong to at least one category, but the categories are not mutually exclusive. A document may be indexed under more 
than one subject category. In spite of this fact, and again to facilitate the computations, the categories are treated, in fact, as exclusive. 

In equation (7) Maron puts forward an equation which contains "attribute numbers" that serve as weights modeling the relevance of a clue word to a category. These weights are derived from the empirical statistics and used to predict the classification.

**Classify document in group 2 using keywords and compare results**
Ultimately Maron presented his findings showing that there is a statistical relationship between clue words and categories. He found that the correct category could be assessed 51% of the time and that the top three predictions would contain the correct classification 80% of the time.

## Borko (1962) - Selecting Classifications using Multidimensional Vectors

In 1962, Borko [published](https://dl.acm.org/doi/abs/10.1145/1460833.1460865) *The Construction of an Empirically Based Mathematically Derived Classiication System* to demonstrate that the classification categories themselves could be derived by means of factor analysis.

From the unique set of terms within the document corpus Borko selects 90 words to use as index terms. These are the most frequend words which occurred 20 or more times, excluding syntactical terms, such as, and, but, of, etc.

Borko constructs a document term matrix to count the occurances of each term within each document. 

In section 2.4 he details how he then constructs a correlation matrix. In this correlation matrix, each cell contains a correlation coefficient between two index terms. The metric is intended to explain how the frequency (occurance count) between documents fluctuates. In other words, it tries to quantify the degree to which these terms occur at the same frequency within the same documents.

**Note**: For a more detailed discussion regarding correlation see the [Correlation Notebook](../../../../Mathematics/Statistics/Covariance%2C%20And%20Correlation.ipynb)


This correlation matrix is then factor analyzed and the term frequency correlations are mapped to the factor space (i.e. a set of orthogonal factors). The factors thus capture term-term relationships accross multiple sets of terms.

**Note** This conceptual space is critically important to understand as it represents a major breakthrough which will be exploited in later advancements (e.g. Deerwester et. al.). The first important property is that the factors are intended to be orthoganal and explain a portion of the observed variance/covariance in the system. Because of this orthoganality, these factors, like the principle axes or components of PCA, can be thought of as cartesian or geometric axeses for a dimension. In other words, the factor space is geometric in nature. Because the EFA process defines a transformation from the observed correlation space to the factor space, every correlation value maps into the factor space coordinate system. An inherant property of the coordinate system is that similar values are geometrically closer to eachother. Again, this is a major advancement that will be exploited in later works.

**Note**: While the technique of using a multidimensional vector to represent semantic meaning is similiar conceptually to that of word embeddings, these two techniques are very different under the hood and arose from different fields of study. For more information see the complementary [Word Embeddings Notebook](Word%20Embeddings.ipynb).

As the factors effectively represent the topics or classes, It becomes difficult to understand what the topics actually are. For this the authors introduce **factor rotation**.

> Factor rotation is designed to locate the 
reference axes in a grid system such that each 
factor can be meaningfully interpreted. The 
interpretation must be made by the investigator 
and is based upon his knowledge of the analytic 
procedures and the subject matter. There is, 
therefore, a degree of subjectivity in the
> selected for each factor. These names may be regarded as hypotheses about the factor names

The rotation is based on **Factor Loading** which is the correlation between the term and the factor; i.e. how much variance in the frequency of the term being observed is matches by variance of the factor.

The rotator (i.e. the human) must adjust the spatial location of the axes so that the factor loading of specific terms is such that an inference can be made. In the paper they walk about terms like boys, girls, school, and reading are heavily loaded while college is not. The authors argue that this suggest the topic that the factor represents was related to non-collegic studdies, perhaps elementary school.

This marks a major shortcoming of the approach.

The first 10 eigenvectors (i.e. those that explain or vapture the most variance/covariance) were selected as factors. This selection inherently implies that a majority of the meaningful variance/covariance is explained by the selected factors. These eigenvectors then mapped back to the intex terms which are treated as the categories for the classification system.

Comparing the index terms with those selected by humans, the author notes that while the terms are different, they are "compatible" (i.e. similar or meaningful enough).

## Borko and Bernick (1963, 1964) - Automation of Document Classification

Continuing Borko's work from the prior year, in 1963, Borko and Bernick [published](https://dl.acm.org/doi/pdf/10.1145/321160.321165) *Automatic Document Classification* followed by a second publication, *Autolnatic Document Classification Part II. Additional Experiments* which was [published](https://dl.acm.org/doi/pdf/10.1145/321217.321219) in 1964.

The purpose of the papers was to conduct experiments to prove that automatic document classification is possible. While Borko (1962) had shown that categories could be selected automatically, he had not gone as far as implementing a system to perform the classification once the categories are chosen.

In the first paper, Borko writes:

>  the experiment is 
designed to demonstrate the relationship of the classification system and the 
procedures for automatic document classification. It is our thesis that if it were 
possible to derive a set of categories which would provide a best possible description of a domain of documents the task of classification would be silnple, and in fact could be done automatically

Borko starts the experiment by using the same corpus of documents as Maron (1961). Borko however, as we will discuss, selects different categories than Maron. He notes that this is advantageous:

> By selecting for our research the same corpus of documents that Maron used, we are provided with an opportunity to compare classification categories as well as automatic indexing procedures. 

As aluded to here, Borko fully automates Maron's classification process (which was partially automated). Specifically, Borko uses an automated method for performing classification he proposed previously in Borko (1962) which relies on factor analysis rather than human judgement. In this new process Borko needed to specify the number of factors to produce from the factor analysis. He reasoned that between 15 and 32 would be sufficient based on an analysis of the corpus, but he did not detail that effort.

**Note**: There is a major caveat here. While Borko et. al. are able to perform classification automatically based on factors, as noted in my analysis of Borko (1962), the rotation of the factors is still a manual and subjective process. In order to establish the orientation of the factor axes, a human must interpret and augment the factor loadings.

While the process of choosing the categories has changed, the Borko continues to use the same 90 clue words from Maron (1961). Recall that the clue words are used to classify document containing them into the relevant categories.

Another change to the original approach we the adoption of a new predictive formula. Ultimately, because the classification was based on eigenvalues and the prediction of the classification was so too dependent on eigenvalues, the formula needed to change.  According to borko, 

> The final prediction formula was a function of the sum of the products of the normalized factor loadings and the index 
term.

At the end of the paper, Borko et. al. discuss the results. We see that Maron's results were better for both group 1 and group 2. However, because the classification mechanism was optimized based on the data in group 1, the quality of the classification between groups 1 and 2 fell considerably. While Borko's metric also fell, it did so less dramatically indicating less bias. In group two, the two methods successfully classified the same number of documents.

At this point in time, no author had been able to propose a unified and automated approach to both document indexing and document retrieval.

To address this shortcoming, In 1990, Deerwester et.al. [published](http://wordvec.colorado.edu/papers/Deerwester_1990.pdf) "Indexing by latent semantic analysis" which proposes a novel approach to indexing and retrieval of text based resources called "Latent Symantic Indexing".

**Note**: The use of LSI and Latent Semantic Analysis (LSA) seem to be interchangable in the literature.

Not only does it address some of the shortcomings of the previous indexing techniques but it also provides a unified implimentation for retireval which leverages the same underlying framework as the indexing technique.

The authors summarize the additional major differences between their approach and previous works:

> we: (1) examine
problems of reasonable size (1000-2000 document abstracts; and 5000-7000 index terms); (2) use a
rich, high-dimensional representation (about 100 dimensions) to capture term-document relations
(and this appears necessary for success); (3) use a mathematical technique which explicitly
represents both terms and documents in the same space; and (4) retrieve documents from query
terms directly, without rotation or interpretation of the underlying axes and without using
intermediate document clusters

**Note**: The authors mentioned point (1) several times; the size of the dataset and scalability of the solution was of major importance to them.

**Improvements To Indexing**

Previous methods, like Maron (1961) and Borko (1962 - 1964) used algorithms which relied on literal matches of terms between documents in order to quantify relevance or similarity. Based on this measurement, documents were clustered together.

Derrwester et. al. point out howerer that this approach has several shortcomings, like an inability to model words that have multiple meaning, or the inability to model the relationship of a word and a specific document (recall Maron and Borko are cluster based).

Deerwester et. al. propose a more effective model which bypasses some of the shorcomings by allowing additional relationships to be represented.

The paper notes their motivation and philosophy as follows:

> The proposed approach tries to overcome the deficiencies of term-matching retrieval by treating
the unreliability of observed term-document association data as a statistical problem. We assume
there is some underlying latent semantic structure in the data that is partially obscured by the
randomness of word choice with respect to retrieval. We use statistical techniques to estimate this
latent structure, and get rid of the obscuring "noise". A description of terms and documents based
on the latent semantic structure is used for indexing and retrieval.
>
>  By "semantic structure" we mean here only the correlation structure in the way in which individual words
appear in documents; "semantic" implies only the fact that terms in a document may be taken as referents to the
document itself or to its topic


This [paper](https://www.cse.msu.edu/~cse960/Papers/LSI/LSI.pdf) also gives a perspective on their motivations and technique:

> Typically, information is retrieved by literally matching terms in documents with
those of a query. However, lexical matching methods can be inaccurate when they
are used to match a user's query. Since there are usually many ways to express a
given concept (synonymy), the literal terms in a user's query may not match those
of a relevant document. In addition, most words have multiple meanings
(polysemy), so terms in a user's query will literally match terms in irrelevant
documents.
> ...
> Latent Semantic Indexing (LSI) [Deerwester et al] tries to overcome the
problems of lexical matching by using statistically derived conceptual indices
instead of individual words for retrieval. LSI assumes that there is some
underlying or latent structure in word usage that is partially obscured by
variability in word choice.

In the paper, Deetwester et. al. describe the several models considered for modeling the semantic similarity between documents. They then go on to define criteria which the models must satisfy in order to be chosen:

- Adjustable representational richness - Able to support a varying degree of complexity between the relationships of documents. terms, and clusters.
- Explicit representation of both terms and documents
- Computational tractability for large datasets - The algorithm is able to scale with larger data sets

Ultimately they note that a two-mode factor analysis is the only reasonable choice:

> The only model which satisfied all three criteria was two-mode factor analysis... Two-mode factor analysis is a
generalization of the familiar factor analytic model based on singular value decomposition (SVD)... SVD represents both terms and documents as vectors in a space of choosable dimensionality, and
the dot product or cosine between points in the space gives their similarity. In addition, a program
was available that fit the model in time of order $N^2 x k^3$

So rather than transforming the document term matrix into a correlation matrix and then performing an ordinary SVD which models only the correlations between the frequency of occurance of terms, they instead perform two-mode SVD directly on the document-term matrix. This preserves the relationships between terms and between terms and documents.

> The approach is to take
advantage of implicit higher-order structure in the association of terms with documents ("semantic
structure") in order to improve the detection of relevant documents on the basis of terms found in
queries. The particular technique used is singular-value decomposition, in which a large term by
document matrix is decomposed into a set of ca 100 orthogonal factors from which the original
matrix can be approximated by linear combination.

**Note**: I believe the term *ca* is a mathematical shorthand for the latin word *circa* meaning around or approximately. For more information of equivalent abreviations see this [paper](https://www.math.ucdavis.edu/~anne/WQ2007/mat67-Common_Math_Symbols.pdf).

> In this
reduced model all the term-term, document-document and term-document similarity is now
approximated by values on this smaller number of dimensions. The result can still be represented
geometrically by a spatial configuration in which the dot product or cosine between vectors
representing two objects corresponds to their estimated similarity.

**Note**: With the dot product, the angle between the vectors is being used to measure similarity not euclidean distance between the vectors.

With this new indexing approach, more information about the relationships of the documents and terms are preserved. The factors that are produced by the decomposition are able to capture more information in a smaller, or lower-dimensional, space.

> In two-mode factor analysis one begins not with a square symmetric matrix relating pairs of only
one type of entity, but with an arbitrary rectangular matrix with different entities on the rows and
columns, e.g., a matrix of terms and documents. This rectangular matrix is again decomposed into
three other matrices of a very special form, this time by a process called "singular-value decomposition" (SVD). (The resulting matrices contain "singular vectors" and "singular values".)
As in the one-mode case these special matrices show a breakdown of the original relationships into
linearly independent components or factors. Again, many of these components are very small, and
may be ignored, leading to an approximate model that contains many fewer dimensions. In this
reduced model all the term-term, document-document and term-document similarity is now
approximated by values on this smaller number of dimensions. The result can still be represented
geometrically by a spatial configuration in which the dot product or cosine between vectors
representing two objects corresponds to their estimated similarity.

That being said, lower dimensionality is not an a priori concern

> Unlike many typical uses of factor analysis, we are not necessarily interested in reducing the
representation to a very low dimensionality, say two or three factors, because we are not interested
in being able to visualize the space or understand it. But we do wish both to achieve sufficient power
and to minimize the degree to which the space is distorted. We believe that the representation of
conceptual space for any large document collection will require more than a handful of underlying
independent "concepts", and thus that the number of orthogonal factors that will be needed is likely
to be fairly large. Moreover, we believe that the model of a Euclidean space is at best a useful
approximation. In reality, conceptual relations among terms and documents certainly involves more
complex structures, including, for example, local hierarchies and non-linear interactions between
meanings. More complex relations can often be made to approximately fit a dimensional
representation by increasing the number of dimensions. In effect, different parts of the space will be
used for different parts of the language or object domain. Thus we have reason to avoid both very
low and extremely high numbers of dimensions. In between we are guided only by what appears to
work best. What we mean by "works best" is not (as is customary in some other fields) what
reproduces the greatest amount of variance in the original matrix, but what will give the best
retrieval effectiveness.

The vectors produced by the Factor Analysis represent the relationships (term-term, document-document and term-document) in a lower dimentsional space which makes the calculations to address overall similarity more simple.

But this approach does not fix the problem of factor roation. In fact, it bypasses the problem altogether:

> We make no attempt to interpret the underlying factors, nor to "rotate"
them to some meaningful orientation. Our aim is not to be able to describe the factors verbally but
merely to be able to represent terms, documents and queries in a way that escapes the unreliability,
ambiguity and redundancy of individual terms as descript
>
> It is possible to reconstruct the original term by document matrix from its factor weights with
reasonable but not perfect accuracy. It is important for the method that the derived k -dimensional
factor space not reconstruct the original term space perfectly, because we believe the original term
space to be unreliable. Rather we want a derived structure that expresses what is reliable and
important in the underlying use of terms as document referents.ors

The authors also explain that this approach alleviates the issue of a word having multiple meanings because indexing and queries are actually happinging in the decomposed or artificial space created by the SVD technique: 

> The latent semantic indexing methods that we have discussed, and in particular the singular value decomposition technique that we have tested, are capable of improving the way in which we
deal with the problem of multiple terms referring to the same object. They replace individual terms
as the descriptors of documents by independent "artificial concepts" that can be specified by any one
of several terms (or documents) or combinations thereof. In this way relevant documents that do not
contain the terms of the query, or whose contained terms are qualified by other terms in the query or
document but not both, can be properly characterized and identified.
>

**Document Retrieval**

The idea of document retrieval is to enable the user to search or query the datastore for documents of interest. Ideally the index by which the document are catalogued will preserve the relevant metadata and relationships so that the search mechanism can be implimented effectively.

A key advancement of this paper is that the retrieval mechanism leverages the same index as the one produced by the document indexing system. Because of this, a query system can be engineered to search the index and return documents which are relevant to the query.

In order to make this work, the query is transformed into a format that is compatible with the index. Because the index inherently exhibits properties which allow document similarity to be assessed and quantified; transforming the query into a "document-like" format allows the query to be directly compared to documents. The documents that are more similar to the query are considered more relevant. The search system can also set a threshold for how many document are returned as part of the query results and what minimum relevance is required in order for the document to be returned.

>  Documents are represented by ca 100 item
vectors of factor weights. Queries are represented as pseudo-document vectors formed from
weighted combinations of terms, and documents with supra-threshold cosine values are returned.

**Note**: I believe the term *ca* is a mathematical shorthand for the latin word *circa* meaning around or approximately. For more information of equivalent abreviations see this [paper](https://www.math.ucdavis.edu/~anne/WQ2007/mat67-Common_Math_Symbols.pdf).

**Note** According to Meriam-Webster, the term supra-threshold means "of sufficient strength or quantity to produce a perceptible physiological effect". In this context, Deetweter is saying that only documents with a cosine similarity value exceeding some threshold are returned in the query results. We can infer that is the case through several passages such as the one below:


> Generally either a threshold is set for

## Hofmann (1999). Probabilistic LSI

In 1999, while researching at berkley, Hofmann [published](https://web.archive.org/web/20101214074049/http://www.cs.brown.edu/~th/papers/Hofmann-SIGIR99.pdf) a paper called *Probabilistic Latent Semantic Indexing*.

> Although LSA has been applied with remarkable success

## Blei et. al. (2003) - LDA applied to ML

In Januaray 2003, Blei et. al. [published](https://web.archive.org/web/20120501152722/http://jmlr.csail.mit.edu/papers/v3/blei03a.html) *Latent Dirichlet Allocation* in the Journal of Machine Learning Research. This is the earliest application of LDA within the machine learning field that I could find.

According to the abstract: 

> We describe latent Dirichlet allocation (LDA), a generative probabilistic model for collections of discrete data such as text corpora. LDA is a three-level hierarchical Bayesian model, in which each item of a collection is modeled as a finite mixture over an underlying set of topics. Each topic is, in turn, modeled as an infinite mixture over an underlying set of topic probabilities. In the context of text modeling, the topic probabilities provide an explicit representation of a document. We present efficient approximate inference techniques based on variational methods and an EM algorithm for empirical Bayes parameter estimation. We report results in document modeling, text classification, and collaborative filtering, comparing to a mixture of unigrams model and the probabilistic LSI model.

**Note**: In this context the term *EM algorithm* refers to the variational Bayesian Expectation–Maximization algorithm. This is closely related to the Maximum Liklihood Estimation (MLE) technique. MLE gives you the objective to maximize. EM is one method for finding this maximum. In other words, EM finds (or attempts to find) the MLE iteratively. There are other methods besides EM that can be applied to find the MLE.

## Liu et. al. (2017)

Other topic models are generally extensions on LDA, such as Pachinko allocation, which improves on LDA by modeling correlations between topics in addition to the word correlations which constitute topics. 

Hierarchical latent tree analysis (HLTA) is an alternative to LDA, which models word co-occurrence using a tree of latent variables and the states of the latent variables, which correspond to soft clusters of documents, are interpreted as topics.
