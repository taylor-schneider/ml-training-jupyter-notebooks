# Overview

The term **atomic** has several meanings depending on the field of study. 

In the case of prompt engineering, the term atomic draws its meaning from work in mathematical logic and analytical philosophy that was conducted in the 1960's and 70's. The basic premise was that complex statements or questions could be broken down into the smallest units of thought which would be easier to assess as true or false. The larger system could then agregate the smaller assessments of truth values to establish a value for the agregate.

[Wikipedia has a nice summary of the concept](https://en.wikipedia.org/wiki/Atomic_sentence): 

> An atomic sentence is a type of declarative sentence which is either true or false (may also be referred to as a proposition, statement or truthbearer) and which cannot be broken down into other simpler sentences. For example, "The dog ran" is an atomic sentence in natural language, whereas "The dog ran and the cat hid" is a molecular sentence in natural language.
>
> [wiki - atomic sentance](https://en.wikipedia.org/wiki/Atomic_sentence)

We see that other terminology from physics is also incorporated into this analog; molecular sentances are thus sentences which contain multiple pieces of atomic information that could in turn be reduced into atomic sentances.

In the context of using LLMs for retrieval tasks, we see this concept of atomic is borrowed and used to distill information.

We can see examples of this, in this [article about atomic fact decomposition](https://arxiv.org/html/2410.16708v1), the authors discuss decomposing long form generated responses into molecular clauses and then into atomic facts. The premise of the idea is that these decomposed attomic components can then individually be used to retrieve relevant information which can then be used to compose a more relevant and informationally rich response.

In the case of prompt engineering, we see the term being use to qualify prompts that are constructed using acomic sentences or facts.

>An Atomic Prompt is a concise, self-contained instruction or question given to an AI model that focuses on a single, specific task or concept. In the context of Generative AI, atomic prompts are designed to elicit targeted responses without requiring additional context or follow-up prompts. These prompts are typically short, clear, and precise, allowing for more efficient and accurate outputs. Atomic prompts are often used as building blocks for more complex interactions or to create consistent, reusable components in AI-powered applications and workflows.
>
>
>Examples of atomic prompts include:
>"Write a haiku about the ocean."
>"Translate 'Good morning' to French."
>"List three benefits of regular exercise."
>
>These prompts are straightforward, focused, and leave little room for misinterpretation, ensuring that the model's response is directly aligned with the user's intent.
>
> More complex prompts that contain multiple instructions, examples, etc. are usually referred to as Compound Prompts.
