# Intuition and Etymology

In general, a prompt is an event, question, or action which in turn triggers a corresponding response. We see prompts every day: when reporters ask questions during interviews; then the parking garage requires a paid ticket to allow a vehicle to exit; or when we a person decides to take a walk because the weather is nice.

If we restrict this general definition to apply within the context of NLP, specifically LLMs, a prompt is a structured input sequence which is designed to produce a specific output sequence. And prompt engineering is the process of deliberately refining and systematically improving the quality of the generated output sequence given the specific objective of the prompt.

Traditionally, users of NLP models would need to use the same structure for their prompt as was provided to the model when it was trained. However, as we will see, emergent abilities allow this restriction to be relaxed and allow the users to define the manner in which they will interact with the LLM.

According to [Liu and Chilton (2023)](https://arxiv.org/abs/2109.06977),

> The term prompt engineering originally came from a popular post online about GPT3 (a
large language model) and its capabilities for writing creative fiction.
The author, Gwern Branwen [5], suggested that prompt engineering models could become a new paradigm for interaction; users
need only figure out how to prompt a model to elicit the specific
knowledge and abstractions necessary for completing downstream
tasks.

Reviewing Brawen's [cited blog post](https://gwern.net/gpt-3) from February 2022, we see that Brawen was researching how to get better performance when interacting with LLms, specifically the GPT series models (specifically GPT-3) at the time the article was written. 

We see Brawen breifly describe how prompt engineering came about, attributing the birth of the practice as a reaction to the discovery of the emergent ability phenomenon which was being demonstrated by the GTP series models:

> The GPT-3 neural network is so large a model in terms of power and dataset that it exhibits qualitatively different behavior: you do not apply it to a fixed set of tasks which were in the training dataset, requiring retraining on additional data if one wants to handle a new task (as one would have to retrain GPT-2); instead, you interact with it, expressing any task in terms of natural language descriptions, requests, and examples, tweaking the prompt until it “understands” & it meta-learns the new task based on the high-level abstractions it learned from the pretraining.This is a rather different way of using a DL model, and it’s better to think of it as a new kind of programming, prompt programming, where the prompt is now a coding language which programs GPT-3 to do new things.

## The Four Horsemen

In his Brawen's [blog post](https://gwern.net/gpt-3), he elaborates on what he calls the "Four Horsemen" of Generative AI: 
> The 4 Horsemen: short context, bad prompts, BPEs, random sampling. My rule of thumb when dealing with GPT-3 is that if it is messing up, the errors are usually attributable to one of 4 problems: too-short context windows, insufficient prompt engineering, BPE encoding making GPT-3 ‘blind’ to what it needs to see to understand & solve a problem, or noisy sampling sabotaging GPT-3’s attempts to show what it knows. Another useful heuristic is to try to express something as a multi-step reasoning process or “inner monologue”, such as a dialogue: because GPT-3 is a feedforward NN, it can only solve tasks which fit within one “step” or forward pass; any given problem may be too inherently serial for GPT-3 to have enough ‘thinking time’ to solve it, even if it can successfully solve each intermediate sub-problem within a step. So people have demonstrated that GPT-3 won’t solve a simple math problem in a single step, but it will solve it if you reframe it as a ‘dialogue’

## Emergent Abilities

Due to the inherent architecture of a neural network, the size (the number of neuron weights that can be trained) of an LLM can scale indefinitely. As such, one would normally expect a linear relationship between the models size and its performance.

As LLMs increase in size it is possible that there exists a non-linear relationship between size an performance resulting in unpredicted increases in performance relative to the increase in size. It's possible that at a certain size, the model "breaks" or "breaks through" the scaling laws that had been observed up until that point and is able to achieve a massive massive and disproportionate increase in a given performance metric.

According to this [wikipedia article](https://en.wikipedia.org/wiki/Large_language_model#Emergent_abilities), some data scientists, for example [Schaeffer et. al. (2023)](https://arxiv.org/abs/2304.15004), argue that the emergent abilities are not unpredictably acquired, but predictably acquired according to a smooth scaling law. This specific area of research is still open to debate.

This phenomenon is well documented by [Wei et al. (2022)](https://arxiv.org/abs/2206.07682). Figure 2 below demonstrates this phenomenon visually. The break can be mathematically found by looking at a log-log plot (a two dimensional plot where the x and y-axis follow a logarithmic scale rather than the standard unit scale) for a point where the slope of the line changes.

The emergent ability is the ability that experiences the break; thus the term is used to refer to an ability derived from being scaled to the point where a break was observed.

Its possible that a model might posess several different emergent abilities and its also possible there exist multiple break outs of performance. This fact is often highlighted in discussions about the evolution of prompt engineering.

Another important aspect of emergent abilities is that an ability might emerge that the model was not explicitly trained to perform. Prompt engineering for example, is made possible because of the emergent abilities phenomenon. The LLMs in vogue are able to acquire multiple emergent abilities that allows them to perform a wide range of tasks despite the fact that they were not explicitly trained to perform those specific tasks.

In the case of prompt engineering, the user is able to craft their own prompt which the model is able to respond to reasonably well, rather than being limited to using only the prompts created by the data scientist that were used during the training session.

### In-Context Learning

The term in-context learning refers to model's ability to learn from the prompt. With this ability, information in the prompt will be able to affect the specific response being generated. The user can inject facts, style the output, give examples of the desired syntax, and more.

## Alexander (1977) - Pattern Language

In 1977, Christopher Alexander [published](https://arl.human.cornell.edu/linked%20docs/Alexander_A_Pattern_Language.pdf) a two volume work titled *A Pattern Language*.

> This book (Volume 1) provides a language, for building and planning; the other book (Volume 2) provides the theory and instructions for the use of the language.





The books intro establishes that the book describes detailed patterns for towns and neighborhoods, houses, gardens, and rooms however many see this work as being relevant to work in digital spaces and not just physical spaces. 

The book attempts to establish reusable patters for building blocks that will ultimately be used to compose larger structures. A key tenant is that these modules need to be designed to serve their user and maximize utility and that the solutions being constructed need to consider the interplay between the different components and so too serve the user. And because of this implicit rule system for building, Alexander proposes that the landuage can become "alive".

> It is shown there, that towns and buildings will not be able to become alive, unless they are made by all the people in society, and unless these people share a common pattern language, within which to make these buildings, and unless this common pattern language is alive itself.

In modern parlance, I believe we would refer to this phenomena as being generative.

His designs have been adapted to several fields including creative problem solving and prompt engineering.

## Perez (2017) - Self-Generating Pattern Lanugage For GPT

Carlos Perez is a data scientist working in the deep learning and AI space. He is familiar with Alexander's work regarding pattern languages and has written several articles and books including *A Pattern Language for Deep Learning* and *A Pattern Language for Generative AI* which extend Pattern Landuage concepts into Perez's respective fields. The latter was written while interacting with the GPT family of models.

Perez acknodleges Alexander's concept of what he called a generative language for architectural design:

> The idea was that as you build this living language, it becomes a language that different architects could use to exchange how basically the motivations for their design.
> 
> [The Jim Rutt Show - EP 187](https://jimruttshow.blubrry.net/the-jim-rutt-show-transcripts/transcript-of-ep-187-carlos-perez-on-a-pattern-language-for-generative-ai/)

According to Perez:

> (Alexander's) ideas where later adopted by Object Oriented Programming (OOP) practitioners to describe the design of OOP programs, these were named Design Patterns. These were extended further into other domains like SOA and High Scalability.
> 
> [A Pattern Language For Deep Learning](https://www.linkedin.com/pulse/pattern-language-deep-learning-carlos-perez/)

> That’s known as the Design and Banners book or the Gang of Four book about 20 years ago or probably a little bit more. I’ve always been a fan of that particular methodology of developing languages for tacit knowledge that exists.
> 
> [The Jim Rutt Show - EP 187](https://jimruttshow.blubrry.net/the-jim-rutt-show-transcripts/transcript-of-ep-187-carlos-perez-on-a-pattern-language-for-generative-ai/)

Perez goes on to describe how he used the GPT-4 model to self-generate prompting templates

> Then one Saturday morning, I had a tweet that queried whether someone had written a pattern language for prompting.
>
> ...
>
>  I’ve always been a fan of that particular methodology of developing languages for tacit knowledge that exists. I was thinking, “Okay. Let’s see if I can apply this to the growing space of prompting that we’re discovering with things like GPT-4.”
>
> But to my surprise, I could actually query GPT-4 for that information and that’s what basically I did. I asked GPT-4 first, I asked it to explain what a pattern language was, which it did correctly. The pattern language template, which consists of a particular structure, the name of the pattern, the context, the competing forces, some examples, and so forth. It has a particular form. Then I asked it generate patterns of prompting, and then it cranked out 10. Then I asked it to generate more, give me another 10.
>
> It kept going until it eventually repeated itself and it got to around, I think, it was around 70 different patterns. I knew of some things that it likely didn’t know so I included those additional ones in the book. The self-generating part is that part where GPT-4 itself was aware of the kinds of prompting patterns that it could actually execute. What’s interesting here is when you see the pattern come out and you see the example, you look at it and you say, “This can’t be right. It can’t possibly do this.”
>
> Unfortunately, when it cranked it out, it was a classification that was I would call pedestrian, something that I didn’t find satisfying. That revealed some limitation in the kinds of concepts that GPT would favor. I think in general, it would favor concepts that would be more commonly known. I didn’t take its advice and instead, basically manually categorized the patterns myself under a framework that was more organic, so to speak.

The importance of self-generated patterns is that they should be optimized to exploit the inherent structures within the model. According to the [primo.ai wiki](https://primo.ai/index.php?title=Prompt_Engineering_(PE)):

> Perez argues that self-generating prompts can enable GPT-4 to learn from its own outputs, generate novel and diverse content, and avoid repetition and bias.
