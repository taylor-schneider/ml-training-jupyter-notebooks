# Overview

In this notebook we will explore the concept of prompts and prompt engineering.

As discussed below, prompt engineering began to emerge in the early 2020s as a direct response to the growing power being unlocked as LLMs continued to scale.

At it's core, an LLM is simply a probabilistic model expressing the most likely word given a set of preceding words. The prompt is simply the input sequence provided to the LLM as the set of preceeding words. The LLM will then use this as a seed to iteratively generate a sequence of a prescribed length.

At a certain point however, the LLM that has become big enough and has been trained on enough of "the right data" will start to accept a wider range of commands (provided through the input sequence) for a wider range of tasks. In other words, the model will start to understand how to both interpret instructions and perform tasks. As such, in the input sequence, the user can not only define the task but also provide information relevant to the completion of the task.

The trick is to construct a useful prompt. This is called Prompt Engineering. There are many use cases for prompt engineering. Some good, such as RAG, some nefarious such as jailbreaking. We will discuss both.

# Intuition and Etymology

In the abstract sense of the word, a prompt is an event, question, or action which in turn triggers a corresponding response. We see prompts every day: when reporters ask questions during interviews; then the parking garage requires a paid ticket to allow a vehicle to exit; or when we a person decides to take a walk because the weather is nice.

If we restrict this general definition to apply within the context of NLP, specifically LLMs, a prompt is a structured input sequence which is designed to produce a specific output sequence. And prompt engineering is the process of deliberately refining and systematically improving the quality of the generated output sequence given the specific objective of the prompt.

Traditionally, users of NLP models would need to use the same structure for their prompt as was provided to the model during the training seuence. And this was generally due to the fact that a model was intended to perform a speciic function. For example, when searching a search engine, the input is expected in a certain format.  

However, as we will see, emergent abilities allow models to perform tasks they were not explicitly trained to perform. This phenomenon gives greater power to the user; through the manipulation of the input sequence, a user can instruct model behavior which might be non-standard or unsepected. Conversely however, it also leads to the possibility that a model may confuse or misinterpret instructions.

All of these considerations lead to the emergence and evolution of prompt engineering.

According to [Liu and Chilton (2023)](https://arxiv.org/abs/2109.06977),

> The term prompt engineering originally came from a popular post online about GPT3 (a
large language model) and its capabilities for writing creative fiction.
The author, Gwern Branwen [5], suggested that prompt engineering models could become a new paradigm for interaction; users
need only figure out how to prompt a model to elicit the specific
knowledge and abstractions necessary for completing downstream
tasks.

Reviewing Brawen's [cited blog post](https://gwern.net/gpt-3) from February 2022, we see that Brawen was researching how to get better performance when interacting with LLms, specifically the GPT series models (specifically GPT-3) at the time the article was written. 

We see Brawen breifly describe how prompt engineering came about, attributing the birth of the practice as a reaction to the discovery of the emergent ability phenomenon which was being demonstrated by the GTP series models:

> The GPT-3 neural network is so large a model in terms of power and dataset that it exhibits qualitatively different behavior: you do not apply it to a fixed set of tasks which were in the training dataset, requiring retraining on additional data if one wants to handle a new task (as one would have to retrain GPT-2); instead, you interact with it, expressing any task in terms of natural language descriptions, requests, and examples, tweaking the prompt until it “understands” & it meta-learns the new task based on the high-level abstractions it learned from the pretraining.This is a rather different way of using a DL model, and it’s better to think of it as a new kind of programming, prompt programming, where the prompt is now a coding language which programs GPT-3 to do new things.

## LLM's Ability To Perform Wide Range Of NLP Tasks

Because of their size (i.e. the number of parameters or weights in its neural networks) an LLM can learn a huge set of statistical relationships between terms found in the corpus. And because of the attention mechanism built into the Transformers, which conditions these relationships based on context, the LLM is able to accurately model multiple different relationships between the same set of words and selectively generate using a statistical relationship that is most relevant to the input sequence.

This feature can be exploited and lies at the heart of an LLM's flexability and versatility. If trained on a sufficiently large and diverse corpus of text (text containing summaries, translations, categorizations, songs, etc.), the LLM's size will allow the model to learn to perform a wide range of NLP tasks rather than needing a single model for each specific task was was the case in the previous era of the small language models.

## Temperature

In the context of prompt engineering, the concept of Temperature often arises. Generally speaking, Temperature is a hyperparameter of the LLM which regulates the "creativity" of the model during generation. 

When the LLM is generating the next token, we can think of this as being performed by a monte carlo-esque process which takes a random sample from a distribution. If we play with the moments of the distribution, we can affect how "random" the sample is; thus how random the generated sequence is. By adjusting the temperature we are adjusting how likely the allowed responses can be. A lower temperature usually results in a more deterministic process (less randomness) while a higher temperature results in a more stochastic process (more randomness).

Lower temperatures make the model increasingly confident in its top choices, while temperatures greater than 1 decrease confidence. 0 temperature is equivalent to argmax/max likelihood, while infinite temperature corresponds to a uniform sampling. [source](https://towardsdatascience.com/how-to-sample-from-language-models-682bceb97277)

In some cases, the temperature and other hyperparameters of the LLM can be adjusted at prompt time. It's not uncommon for the prompt engineer to try and tune these hyperparametes to best suit a particular prompt.

## Top-K (Nucleus) Sampling

Top-k sampling is a technique is used prevent the generative model from going off topic by removing elements from the sampling set which do not cross a threshold of relevance or likelihood. Additionally it allows the model to be creative by allowing for more than one possible word to be considered during decoding.

For more on this topic see [Holtzman et al. (2019)](https://arxiv.org/abs/1904.09751).

## Emergent Abilities

Due to the inherent architecture of a neural network, the size (the number of neuron weights that can be trained) of an LLM can scale indefinitely. As such, one would normally expect a linear relationship between the models size and its performance.

As LLMs increase in size it is possible that there exists a non-linear relationship between size an performance resulting in unpredicted increases in performance relative to the increase in size. It's possible that at a certain size, the model "breaks" or "breaks through" the scaling laws that had been observed up until that point and is able to achieve a massive massive and disproportionate increase in a given performance metric.

According to this [wikipedia article](https://en.wikipedia.org/wiki/Large_language_model#Emergent_abilities), some data scientists, for example [Schaeffer et. al. (2023)](https://arxiv.org/abs/2304.15004), argue that the emergent abilities are not unpredictably acquired, but predictably acquired according to a smooth scaling law. This specific area of research is still open to debate.

This phenomenon is well documented by [Wei et al. (2022)](https://arxiv.org/abs/2206.07682). Figure 2 below demonstrates this phenomenon visually. The break can be mathematically found by looking at a log-log plot (a two dimensional plot where the x and y-axis follow a logarithmic scale rather than the standard unit scale) for a point where the slope of the line changes.

The emergent ability is the ability that experiences the break; thus the term is used to refer to an ability derived from being scaled to the point where a break was observed.

Its possible that a model might posess several different emergent abilities and its also possible there exist multiple break outs of performance. This fact is often highlighted in discussions about the evolution of prompt engineering.

Another important aspect of emergent abilities is that an ability might emerge that the model was not explicitly trained to perform. Prompt engineering for example, is made possible because of the emergent abilities phenomenon. The LLMs in vogue are able to acquire multiple emergent abilities that allows them to perform a wide range of tasks despite the fact that they were not explicitly trained to perform those specific tasks.

In the case of prompt engineering, the user is able to craft their own prompt which the model is able to respond to reasonably well, rather than being limited to using only the prompts created by the data scientist that were used during the training session.

### In-Context Learning

The term in-context learning refers to model's ability to learn from the prompt. With this ability, information in the prompt will be able to affect the specific response being generated. The user can inject facts, style the output, give examples of the desired syntax, and more. 

An example would be to ask a model to summarize a piece of text; both the text itself and the instruction to summarize are learned during the processing of the prompt. Additionally we might provide example of the desired output which can also be learned in-context.

#### N-Shot Learning

With normal supervised learning, we train a model to perform a task given a set of examples. For example, with classification tasks, we might show a model different pictures of animals and have the model learn to classify those animals into groups (eg. mammal, fish, insect). The assumption being that when the model is "online" and making predictions, it will be classifying animals that were included as part of the training set.

It's possible however, that a model would be able to classify an animal into the correct group without having been explicitly trained how. Instead, through the emergent ability of in-context learning, the model may be given information through the prompt which will help it perform the classification.

There are different approaches and types of N-shot learning as we will see. The practice serves as a solution to problems where large or complete training sets are impractical or impossible.

##### **One-Shot Learning**

One-shot learning is the practice of providing a model with a single example of a new task being performed correctly as part of the input to the inference mechanism, with the expectation that the model will be able to in-context learn how to perform the specific task it was not trained to perform.

Continuing the example of classifying animals, If we were to provide the model with a single picture of a mushroom telling it that this organism should be classified as a fungi, and then have the model perform the classification of another image (which could be a fungi), this would be one-shot learning. The model is given one-shot to in-context learn what a fungi is.

Adding multiple examples is how we get two-shot, three-shot, "few-shot", and ultimately N-shot learning.

##### **Zero-Shot Learning (ZSL)**

Zero-shot learning is when a model is one-shot learning, but rather than using complete examples it is using incomplete examples. Continuing the example, rather than being given a photo of a fungi, which completely illustrates all the important characteristics, it might only be given a partial example which only contains a subset of the defining characteristics. Thus the model would be able to identify a new genus which it has never seen.

Additionally, with more complicated models, it might be the case that the model is only given text as the context, and never actually "sees" a photo of the characteristics being used to make the classification.

For this reason the technique is called zero-shot; because zero examples were provided to the model through the inference context.

According to [wikipedia](https://en.wikipedia.org/wiki/Zero-shot_learning):

> The first paper on zero-shot learning in natural language processing appeared in 2008 at the AAAI’08, but the name given to the learning paradigm there was dataless classification.[3] The first paper on zero-shot learning in computer vision appeared at the same conference, under the name zero-data learning.[4] The term zero-shot learning itself first appeared in the literature in a 2009 paper from Palatucci, Hinton, Pomerleau, and Mitchell at NIPS’09.[5] This terminology was repeated later in another computer vision paper[6] and the term zero-shot learning caught on, as a take-off on one-shot learning that was introduced in computer vision years earlier.[7]

## Alexander (1977) - Pattern Language

In 1977, Christopher Alexander [published](https://arl.human.cornell.edu/linked%20docs/Alexander_A_Pattern_Language.pdf) a two volume work titled *A Pattern Language*.

> This book (Volume 1) provides a language, for building and planning; the other book (Volume 2) provides the theory and instructions for the use of the language.





The books intro establishes that the book describes detailed patterns for towns and neighborhoods, houses, gardens, and rooms however many see this work as being relevant to work in digital spaces and not just physical spaces. 

The book attempts to establish reusable patters for building blocks that will ultimately be used to compose larger structures. A key tenant is that these modules need to be designed to serve their user and maximize utility and that the solutions being constructed need to consider the interplay between the different components and so too serve the user. And because of this implicit rule system for building, Alexander proposes that the landuage can become "alive".

> It is shown there, that towns and buildings will not be able to become alive, unless they are made by all the people in society, and unless these people share a common pattern language, within which to make these buildings, and unless this common pattern language is alive itself.

In modern parlance, I believe we would refer to this phenomena as being generative.

His designs have been adapted to several fields including creative problem solving and prompt engineering.

## Lester et al. (2021) - Prompt Tuning

On April 18th 2021, Lester et al. [published](https://arxiv.org/abs/2104.08691) *The Power of Scale for Parameter-Efficient Prompt Tuning*.

This paper is widely cited as the birth of prompt tuning and soft prompts.

The basic idea put forward is that rather than have a developer create "hard prompts" consisting of static text based templates, instead, a data scientist could train an intermediary model to produce the best prompt for a given task. This model would then generate a prompt dynamically, based on the user's request, what would then be fed to the LLM.

## Wei et al. (2022) - Chain-Of-Thought Prompting

On January 22nd 2026, Wei et al. [published](https://arxiv.org/abs/2201.11903) *Chain-of-Thought Prompting Elicits Reasoning in Large Language Models*. In the paper, the authors explore how an LLM's emergent ability to perform complex reasoning tasks is improved by generating a chain of thought—a series of intermediate reasoning steps. 

As we will see, Chain-Of-Thought prompting is a prompting technique which structures the prompt such that it contains examples of a question, a chain-of-thought (i.e. a series of reasoning steps which lead to the ultimate conclusion), and an answer. The authors describe the prompt format as being composed of triples: <input, chain of thought, output>.

The technique was inspired by two main areas of prior research:
1. It had been shown that an LLMs ability to perform arithmetic reasoning can benefit from generating natural language rationales that lead to the final answer
2. It had been shown that LLMs posess the emergent ability to perform in-context few-shot learning via prompting.

These two techniques however each had their own limitations. The authors hypothesized that applying both of them in a combined technique would yield a hybrid approach which would not suffer from the same limitations.

In the abstract, the authors give examples of few-shot prompting using standard and chain-of-thought prompting schemas:

>

Here we can see that the user's question is embedded within a prompt structured to provide an example and then provide a restructured example of the user's input. Reformatting this as a pseudo-code template we might imagine the prompt template to be:

## Yao et al. (2022) - ReAct Prompting (Agentic RAG)

On October 6th 2020, a team working at Google Brain [published](https://arxiv.org/abs/2210.03629) *ReAct: Synergizing Reasoning and Acting in Language Models*.

The paper presents a new prompting paradigm which aims to combine and enhance an LLMs ability to reason and to act.

> ReAct performs more than just isolated, fixed reasoning, and integrates
model actions and their corresponding observations into a coherent stream of inputs for the model to
reason more accurately and tackle tasks beyond reasoning (e.g. interactive decision making).

The basic idea is to make use of verbal reasoning traces and actions to enhance the problem solving stills of LLMs. The basic workflow is broken down into two key patterns: **reason to act**, and **act to reason**. 

> ReAct prompts LLMs to generate both verbal reasoning traces and actions pertaining to a task in an
interleaved manner, which allows the model to perform dynamic reasoning to create, maintain, and
adjust high-level plans for acting (reason to act), while also interact with the external environments
(e.g. Wikipedia API) to incorporate additional information into reasoning (act to reason).

A concrete implimentation for the ReAct pattern is not provided. However, while not explicitly stated, we can infer from the paper, and from looking at code examples ([here](https://towardsai.net/p/machine-learning/finetuning-llms-for-react) and [here](https://towardsdatascience.com/using-langchain-react-agents-for-answering-multi-hop-questions-in-rag-systems-893208c1847e)) that the two patterns are chained together as successive steps inside an orchestrated loop; which I am referring to as the "ReAct Loop" in the diagram below. The LLM will reason which action needs to execute, and then an agent will parse the LLM's output and execute the coresponding action. This is broken down into the thought-act-observation sequence.

The action may be to terminate the ReAct loop because a task has successfully completed or failed; or it may be to execute some integration with an external system in order to make a relevant observation for the next iteration of the loop. Additionally, the agent is responsible to orchestrating the ReAct process; it is reponsible for generating the initial prompt, instructing the llm to identify the next action, executing the action, and iterating over the loop. This process is visualized in the diagram below.

I was able to figure this out by playing with the [langchain example](https://python.langchain.com/v0.1/docs/modules/agents/agent_types/react/) of using their ReAct Agent and stepping into the code with a debugger. I found that the langchain framework is quite complex and not so straight forward, however I was able to see the agent creating a PromptTemplate, send input to the LLM, and then parse the output. The ReAct prompts are implimented in a manner similar to the Chain-Of-Thought template (see [Wei (2022)](#Wei-(2022)---Chain-Of-Thought-Prompting)) where the LLM is provided a repetative set of few-shot exmples which are formatted to differentiate a question from the reasoning steps used to generate the final answer.



An example ReAct prompt template might be as follows (note the comments denoted by '#' are jsut there to clarify the sections and are not actually part of the template):

```
# Primary instruction
You are a ReAct Solver. Solve or answer the provided question or task.

# ReAct Instruction
Solving the problem requires interleving the Thought, Action, and Observation steps as shown in the few-shot example below.

# Action space definitions
An action can be selected from the following list:

(1) Search[entity] - which returns the first 5 sentences from the corresponding entity wiki page if it exists, or else suggests top-5 similar entities from the Wikipedia search engine.

(2) Lookup[keyword] - which would return the next sentence in the page containing string, simulating Ctrl+F functionality on the browser.

(3) Finish[Answer] - which returns the answer and finishes the task.

# Few-Shot Examples
Below is an example of a successful ReAct solver session:

Question: What profession does Nicholas Ray and Elia Kazan have in common?
Thought 1: I need to search Nicholas Ray and Elia Kazan, find their professions, then
find the profession they have in common.
Action 1: Search[Nicholas Ray]
Observation 1: Nicholas Ray (born Raymond Nicholas Kienzle Jr., August 7, 1911 - June 16,
1979) was an American film director, screenwriter, and actor best known for
the 1955 film Rebel Without a Cause.
Thought 2: Professions of Nicholas Ray are director, screenwriter, and actor. I need
to search Elia Kazan next and find his professions.
Action 2: Search[Elia Kazan]
Observation 2: Elia Kazan was an American film and theatre director, producer, screenwriter
and actor.
Thought 3: Professions of Elia Kazan are director, producer, screenwriter, and actor.
So profession Nicholas Ray and Elia Kazan have in common is director,
screenwriter, and actor.
Action 3: Finish[director, screenwriter, actor]

# User Question

Question: {user_question}
```

The source code for the prompt can be found [here](https://smith.langchain.com/hub/hwchase17/react). The prompt was as follows:

> Answer the following questions as best you can. You have access to the following tools:
>
>{tools}
>
>Use the following format:
>
>Question: the input question you must answer </br>
>Thought: you should always think about what to do </br>
>Action: the action to take, should be one of [{tool_names}] </br>
>Action Input: the input to the action </br>
>Observation: the result of the action </br>
>... (this Thought/Action/Action Input/Observation can repeat N times) </br>
>Thought: I now know the final answer </br>
>Final Answer: the final answer to the original input question
>
>Begin!
>
>Question: {input}
>
>Thought:{agent_scratchpad}

A major different however is that the ReAct reasoning process is not limited to a single turn or iteration and instead is allowed to unfold over multiple attempts in a kind of loop-like fashion. The paradigm basically bounces back and fourth between the two patterns: it reasons to determine which act is appropriate, it then acts to updatethe state so that it can again reasion. This may happen several times until the algorithm either fails or succeeds. 

A ReAct prompt template contains the following parts:
1. Primary Instruction - The actual task the LLM is being instructed to perform. For example, solve a problem or answer a question.
2. ReAct Instruction - The instruction informing what the major steps in the chain-of-thought process (e.g. thought, action, observation).
3. Action Space Definitions - The set of actions that the LLM can choose to execute if it does not know the answer.
4. Few-Shot Examples - A set of examples of successfully performing ReAct
5. The user's question formatted in the ReAct way.

The authors also note that finetuning an LLM on ReAct prompts also helps the overall performance when performing in-context few-shot learning.

This  the magic of the LLM allows it to understand that is needs to treat the keywords "entity," "keyword," and "answer" in the action defintions as placeholders which will be filled in during runtime. 

**Note**: I assume that this ability is due to the LLM being trained on a corpus of data which contained examples featuring a similar syntax. The "best" prompt / action syntax is still an open question at this point in time.

The authors summarize their contibutions in the paper as:

> 1.  We introduce ReAct, a novel promptbased paradigm to synergize reasoning and acting in language models for general task solving; 
> 2.  We perform extensive experiments across diverse benchmarks to showcase the advantage of ReAct in a
few-shot learning setup over prior approaches that perform either reasoning or action generation in
isolation; 
> 3. We present systematic ablations and analysis to understand the importance of acting in
reasoning tasks, and reasoning in interactive tasks; 
> 4. We analyze the limitations of ReAct under the
prompting setup (i.e. limited support of reasoning and acting behaviors), and perform initial finetuning
experiments showing the potential of ReAct to improve with additional training data. Scaling up
ReAct to train and operate on more tasks and combining it with complementary paradigms like
reinforcement learning could further unlock the potential of large language models.

The authors also note:

> ReAct overcomes prevalent issues of hallucination and error propagation in chain-of-thought reasoning by ...  generating human-like task-solving trajectories that are more interpretable than baselines without reasoning traces.


Additionally the authors note that ReAct outperforms several reinforcement learning methods.

## Ahn et al. (2022) - SayCan - Connecting LLMs to Robots

On April 4th 2022, researchers at the Robotics at Google lab [pubished](https://arxiv.org/abs/2204.01691) *Do As I Can, Not As I Say: Grounding Language in Robotic Affordances*.

In this paper, the authors postulate that an LLM could be connected to a stateful environment through an intermediary sensor, agent, or actor (in this case, a robot); allowing the LLM to observe and affect the environment as it attempts to interact with and problem solve within the environment over time.

The abstract reads as follows:

> Large language models can encode a wealth of semantic knowledge
about the world. Such knowledge could be extremely useful to robots aiming to act
upon high-level, temporally extended instructions expressed in natural language.
However, a significant weakness of language models is that they lack real-world
experience, which makes it difficult to leverage them for decision making within
a given embodiment. For example, asking a language model to describe how to
clean a spill might result in a reasonable narrative, but it may not be applicable to
a particular agent, such as a robot, that needs to perform this task in a particular
environment. We propose to provide real-world grounding by means of pretrained
skills, which are used to constrain the model to propose natural language actions
that are both feasible and contextually appropriate. The robot can act as the language model’s “hands and eyes,” while the language model supplies high-level
semantic knowledge about the task. 

A major component to this idea is the *Value Function*; which the authors describe as:

> affordance functions that capture the log likelihood that a particular skill will be able to succeed in the current state. 

To be more precise, a value function models the probability that a skill can complete a task given the current state and the language description of the skill. The prescribed probability function is ultimately used by a classifier type model which then selects the best skillset for the given state/task.

**Note**: The term value function and affordance function are use interchangibly here. My understanding is that affordance has a much more specific meaning within the field of robotics.

The authors name their approach as SayCan can describe the solution of the combination of two components:

- Say - An LLM which provides a task-grounding to determine useful actions for a high-level goal. It describes the probability that each skill contributes to completing the instruction.
- Can - The learned affordance functions which provide a world-grounding to determine what is possible to execute upon the plan. It describes the probability that each skill will succeed. The affordance functions make the LLM aware of the current scene, and
constraining the completions to the skill descriptions makes the LLM aware of the robot’s capabilities.

The authors argue that *combining the two provides the probability that each skill will perform the instruction successfully*.

We understand that the authors intend to use prompt engineering (in-context few-shot learning) to constrole the structure of the llm responses however they note that this is not enough to fully and reliably constrain the outputs:

> it can produce inadmissible actions or language that is
not formatted in a way that is easy to parse into individual steps.

To constrain the outputs, the authors propose using the probability scores yielded by the LLM. The central idea being to have the LLM output the probabilities assigned to the fixed outputs.

**Note**: Reading through the paper, it's not made explicitly clear how the action chosen by the LLM is invoked or triggered. It seemed that no matter how the interface was constructed, the LLM would be consulted but not ultimately responsible for execution. Instead, some higher level orchestrater would prompt the LLM, receive its output (the execution plan), and then trigger the execution.
